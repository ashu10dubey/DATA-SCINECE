
### **Q1. What is meant by time-dependent seasonal components?**  
Time-dependent seasonal components refer to **seasonal variations in a time series that change over time** rather than remaining constant.  
- In **standard seasonality**, seasonal effects repeat in a fixed pattern.  
- In **time-dependent seasonality**, the magnitude, duration, or occurrence of seasonal effects change due to external factors.  

For example, **consumer behavior changes over years** (e.g., holiday sales shifting due to economic trends).  

---

### **Q2. How can time-dependent seasonal components be identified in time series data?**  
Time-dependent seasonality can be detected using:  
1. **Time Series Decomposition** – Breaking down a series into trend, seasonal, and residual components.  
2. **Rolling Window Statistics** – Checking if seasonal patterns change over different time periods.  
3. **Fourier Transform (Spectral Analysis)** – Identifying shifts in seasonal frequency over time.  
4. **Wavelet Transform** – Captures local and evolving periodic patterns.  
5. **Visual Inspection** – Plotting seasonality trends over multiple time periods to observe variations.  

---

### **Q3. What are the factors that can influence time-dependent seasonal components?**  
Several external and internal factors can cause time-dependent seasonality, such as:  
- **Economic Conditions** – Recessions can affect seasonal sales cycles.  
- **Market Trends** – Customer preferences evolve over time.  
- **Regulatory Changes** – New laws affecting demand cycles.  
- **Weather and Climate Shifts** – Seasonal variations change due to global warming.  
- **Technological Advancements** – New innovations may alter consumer behavior patterns.  

For example, **Black Friday sales have become more online-driven over time, changing traditional seasonality trends**.

### **Q4. How are autoregression models used in time series analysis and forecasting?**  
Autoregression (AR) models use past values of a time series to predict future values. The model is defined as:  
$$
X_t = c + \phi_1 X_{t-1} + \phi_2 X_{t-2} + \dots + \phi_p X_{t-p} + \epsilon_t
$$
where:  
- $$ X_t  $$ is the current value,  
- $$ X_{t-1}, X_{t-2}, ... $$ are past observations,  
- $$ \phi  $$ are coefficients,  
- $$ \epsilon_t  $$ is white noise.  

**Applications:**  
- **Stock Market Prediction** – Future stock prices depend on past trends.  
- **Sales Forecasting** – Predicting future demand using historical data.  

---

### **Q5. How do you use autoregression models to make predictions for future time points?**  
To forecast using an AR model:  
1. **Fit the Model** – Estimate coefficients \( \phi \) using least squares or Maximum Likelihood Estimation (MLE).  
2. **Check Stationarity** – Use the **Dickey-Fuller test**; if non-stationary, apply differencing.  
3. **Determine Lag Order (p)** – Use **Partial Autocorrelation Function (PACF)** to identify significant lags.  
4. **Make Predictions** – Use the fitted model to forecast future values iteratively.  
5. **Evaluate Model** – Check residuals for randomness using the Ljung-Box test.  

Example: Predicting tomorrow’s stock price based on the past 10 days’ prices.

### **Q6. What is a moving average (MA) model and how does it differ from other time series models?**  
A Moving Average (MA) model uses past forecast errors rather than past values to predict future points:  
$$
X_t = \mu + \theta_1 \epsilon_{t-1} + \theta_2 \epsilon_{t-2} + \dots + \theta_q \epsilon_{t-q} + \epsilon_t
$$
where:  
- $$ \mu  $$ is the mean,  
- $$ \theta $$ are coefficients,  
- $$ \epsilon_t  $$ are white noise errors.  

**Key Differences:**  
| Model Type  | Uses Past Values | Uses Past Errors | Example |  
|------------|----------------|----------------|---------|  
| **AR (p)**  |  Yes          |  No          | Stock Prices |  
| **MA (q)**  |  No          |  Yes          | Demand Fluctuations |  
| **ARMA (p,q)** |  Yes  |  Yes  | Economic Indicators |  

MA models are good for handling sudden shocks but do not capture long-term dependencies.  

---

### **Q7. What is a mixed ARMA model and how does it differ from an AR or MA model?**  
An **ARMA (AutoRegressive Moving Average) model** combines both **AR (p) and MA (q) components**:  
$$
X_t = c + \sum_{i=1}^{p} \phi_i X_{t-i} + \sum_{j=1}^{q} \theta_j \epsilon_{t-j} + \epsilon_t
$$
where:  
- AR component $$ \phi X_{t-p} $$ captures **long-term dependency**,  
- MA component $$ \theta \epsilon_{t-q}) $$ captures **short-term noise corrections**.  

**Differences:**  
- AR models assume dependence on past values.  
- MA models assume dependence on past errors.  
- ARMA captures both **trend-based** and **error-correcting** components.  

**Use Case:**  
Used in **economic forecasting, sales prediction, and signal processing**, where long-term trends and short-term variations matter.
