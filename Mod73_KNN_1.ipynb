### **Q1. What is the KNN algorithm?**  
The **K-Nearest Neighbors (KNN)** algorithm is a **non-parametric, supervised learning algorithm** used for both **classification and regression** tasks. It classifies a new data point based on the majority vote of its **k-nearest neighbors** (for classification) or computes the average of the nearest neighbors’ values (for regression).  

It works as follows:  
1. Choose the number of neighbors \( K \).  
2. Compute the **distance** between the new data point and all other data points in the dataset.  
3. Identify the **K closest neighbors**.  
4. **Classification**: Assign the majority class among K neighbors.  
5. **Regression**: Compute the average of K neighbors' values.  

### **Q2. How do you choose the value of K in KNN?**  
Choosing the right value of **K** is crucial for the model's performance.  

- A **small K (e.g., 1 or 3)** can make the model sensitive to noise, leading to overfitting.  
- A **large K (e.g., 15, 20, or more)** makes the model more generalized but may smooth out important details.  

**Methods to find the best K:**  
- **Cross-validation**: Use **k-fold cross-validation** to test different values of K.  
- **Elbow method**: Plot accuracy/error against K and choose the point where the error stabilizes.  
- **Odd K values**: For classification, choose **odd K** to avoid ties.  

### **Q3. What is the difference between KNN classifier and KNN regressor?**  

| Feature | KNN Classifier | KNN Regressor |
|---------|--------------|--------------|
| Output Type | Discrete (Class labels) | Continuous values |
| Decision Rule | Majority voting among K neighbors | Average of K neighbors |
| Use Case | Classification problems (e.g., spam detection) | Regression problems (e.g., house price prediction) |
| Performance Measure | Accuracy, Precision, Recall, F1-score | Mean Squared Error (MSE), R-squared |

### **Q4. How do you measure the performance of KNN?**  

**For classification:**  
- **Accuracy** = (Correct predictions) / (Total predictions)  
- **Confusion matrix** (Precision, Recall, F1-score)  
- **ROC Curve & AUC Score**  

**For regression:**  
- **Mean Squared Error (MSE)** = $$( \frac{1}{n} \sum (y_i - \hat{y_i})^2 ) $$
- **Mean Absolute Error (MAE)**  
- **R-squared (\( R^2 \)) score**  

### **Q5. What is the curse of dimensionality in KNN?**  
The **curse of dimensionality** occurs when the number of features (**dimensions**) in the dataset increases.  

**Why is it a problem?**  
- As dimensions increase, the data points become **sparse**, making the concept of "nearest neighbors" meaningless.  
- The Euclidean distance becomes **less informative**, reducing the model's performance.  

**How to handle it?**  
- **Dimensionality reduction** using **PCA (Principal Component Analysis)** or **feature selection**.  
- Use **Manhattan distance** or **cosine similarity** instead of Euclidean distance.  

### **Q6. How do you handle missing values in KNN?**  
Handling missing values in KNN is essential since it relies on distance calculations.  

**Methods to handle missing values:**  
- **Imputation**:  
  - Use **mean/median/mode** for numerical/categorical features.  
  - Use **KNN Imputer**, which fills missing values based on the nearest neighbors.  
- **Remove missing rows** (if the percentage of missing values is low).  
- **Predict missing values** using another model.  

### **Q7. Compare and contrast the performance of the KNN classifier and regressor. Which one is better for which type of problem?**  

| Feature | KNN Classifier | KNN Regressor |
|---------|--------------|--------------|
| Performance | Works well when classes are well separated | Works well when data has a smooth trend |
| Robustness to Outliers | Less robust (misclassified due to noise) | More robust (outliers affect averages less) |
| Use Case | Image recognition, medical diagnosis | Stock price prediction, temperature forecasting |
| Decision Boundary | Non-linear and sensitive to K | Averages out local trends |

**Which one is better?**  
- **KNN Classifier** is better for **categorical predictions** (e.g., sentiment analysis).  
- **KNN Regressor** is better for **continuous predictions** (e.g., predicting sales).  

### **Q8. Strengths and Weaknesses of KNN for Classification and Regression**  

**Strengths:**  
 **Simple & Easy to implement**  
 **Non-parametric (no assumption about data distribution)**  
 **Works well on small datasets**  

**Weaknesses & How to Address Them:**  

| Weakness | Solution |
|----------|---------|
| **Computationally expensive** (O(n)) | Use **KD-Tree** or **Ball-Tree** for faster search |
| **Sensitive to irrelevant features** | Feature selection & PCA |
| **Struggles in high dimensions** | Dimensionality reduction |
| **Requires feature scaling** | Normalize data using Min-Max or StandardScaler |

### **Q9. What is the difference between Euclidean distance and Manhattan distance in KNN?**  

| Distance Metric | Formula | When to Use |
|----------------|---------|-------------|
| **Euclidean Distance** | $$( d = \sqrt{\sum (x_i - y_i)^2} ) $$ | Best for **continuous** data with smooth decision boundaries |


**Key Differences:**  
- **Euclidean** measures straight-line distance (good for curved patterns).  
- **Manhattan** measures distance along axes (better for high-dimensional data).  

### **Q10. What is the role of feature scaling in KNN?**  
KNN relies on distance calculations, so feature scaling is **crucial** to ensure fair comparisons.  

**Why is it needed?**  
- If features have different scales, KNN will **bias towards features with larger scales**.  
- Example: If age (scaled 0-100) and salary (scaled 0-1,000,000) are compared, salary dominates distance calculations.  

**How to scale features?**  
1. **Min-Max Scaling** (Scales between 0 and 1)  
   $$
   x' = \frac{x - \min(x)}{\max(x) - \min(x)}
   $$
2. **Standardization (Z-score normalization)**  
   $$
   x' = \frac{x - \mu}{\sigma}
   $$
   (Centers around mean 0, variance 1)  

**Best Practice:**  
- Use **Min-Max Scaling** for **bounded data** (e.g., pixel intensities).  
- Use **Z-score** for **normally distributed** data.
