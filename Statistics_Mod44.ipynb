
**Q1. What is the Probability Density Function (PDF)?**

**A1.**

The Probability Density Function (PDF) is a function that describes the likelihood of a continuous random variable taking on a specific value.

* **Key Characteristics:**
    * Defined for continuous random variables.
    * The total area under the PDF curve over the entire range of the variable is equal to 1.
    * The probability of the random variable falling within a particular interval is given by the area under the PDF curve over that interval.
    * The PDF itself does not represent the probability of the variable taking on a specific value directly. Instead, it represents the probability *density* at that point.

**Q2. What are the types of Probability Distribution?**

**A2.**

There are numerous types of probability distributions, each with its own unique characteristics and applications. Some of the most common ones include:

* **Continuous Distributions:**
    * **Normal Distribution:** Bell-shaped curve, widely used in many fields.
    * **Uniform Distribution:** All values within a given range are equally likely.
    * **Exponential Distribution:** Often used to model waiting times.
    * **Gamma Distribution:** Generalization of the exponential distribution.
    * **Beta Distribution:** Used to model probabilities and proportions.

* **Discrete Distributions:**
    * **Bernoulli Distribution:** Represents a single trial with two possible outcomes (success or failure).
    * **Binomial Distribution:** Represents the number of successes in a fixed number of independent Bernoulli trials.
    * **Poisson Distribution:** Represents the number of events occurring within a fixed interval of time or space.
    * **Geometric Distribution:** Represents the number of trials required to achieve the first success.

**Q3. Write a Python function to calculate the probability density function of a normal distribution with given mean and standard deviation at a given point.**

```python
import math

def normal_pdf(x, mean, std_dev):
    """
    Calculates the probability density function of a normal distribution.

    Args:
        x: The point at which to evaluate the PDF.
        mean: The mean of the normal distribution.
        std_dev: The standard deviation of the normal distribution.

    Returns:
        The probability density at the given point.
    """
    exponent = -0.5 * ((x - mean) / std_dev)**2
    return (1 / (std_dev * math.sqrt(2 * math.pi))) * math.exp(exponent)

# Example usage
x = 0
mean = 1
std_dev = 2
pdf_value = normal_pdf(x, mean, std_dev)
print(f"PDF at x = {x}: {pdf_value}")
```

**Q4. What are the properties of Binomial distribution? Give two examples of events where binomial distribution can be applied.**

**A4.**

**Properties of Binomial Distribution:**

* **Fixed number of trials (n):** The experiment consists of a fixed number of independent trials.
* **Two possible outcomes:** Each trial has only two possible outcomes (success or failure).
* **Constant probability of success (p):** The probability of success remains constant for each trial.
* **Independent trials:** The outcome of each trial does not affect the outcomes of other trials.

**Examples of Binomial Distribution:**

1. **Flipping a coin 10 times and counting the number of heads.** Each flip is a trial, the outcome is either heads (success) or tails (failure), the probability of heads is constant (assuming a fair coin), and the flips are independent.
2. **The number of defective items in a batch of 100 items produced by a machine.** Assuming the probability of a defective item is constant and the production process is independent for each item.

**Q6. Write a Python function to calculate the cumulative distribution function of a Poisson distribution with given mean at a given point.**

```python
import math

def poisson_cdf(k, mean):
    """
    Calculates the cumulative distribution function of a Poisson distribution.

    Args:
        k: The number of occurrences.
        mean: The mean of the Poisson distribution (lambda).

    Returns:
        The cumulative probability up to k occurrences.
    """
    cdf = 0
    for i in range(k+1):
        cdf += math.exp(-mean) * (mean ** i) / math.factorial(i)
    return cdf

# Example usage
k = 3
mean = 2
cdf_value = poisson_cdf(k, mean)
print(f"CDF at k = {k}: {cdf_value}")
```

**Q7. How Binomial distribution different from Poisson distribution?**

**A7.**

| Feature | Binomial Distribution | Poisson Distribution |
|---|---|---|
| **Type of Events** | Discrete events with two possible outcomes (success/failure) | Discrete events occurring randomly over time or space |
| **Number of Trials** | Fixed number of trials (n) | No fixed number of trials |
| **Probability of Success** | Constant probability of success (p) in each trial | Average rate of occurrence (λ) is constant |
| **Typical Applications** | Coin flips, quality control, clinical trials | Number of calls to a call center, number of arrivals at a queue |

**Q8. Generate a random sample of size 1000 from a Poisson distribution with mean 5 and calculate the sample mean and variance.**

```python
import numpy as np

# Generate a random sample
sample_size = 1000
mean = 5
data = np.random.poisson(mean, size=sample_size)

# Calculate sample mean and variance
sample_mean = np.mean(data)
sample_variance = np.var(data)

print(f"Sample Mean: {sample_mean}")
print(f"Sample Variance: {sample_variance}")
```

**Q9. How mean and variance are related in Binomial distribution and Poisson distribution?**

**A9.**

* **Binomial Distribution:**
    - Mean: μ = n * p (where n is the number of trials and p is the probability of success)
    - Variance: σ² = n * p * (1 - p)

* **Poisson Distribution:**
    - Mean: μ = λ (where λ is the average rate of occurrence)
    - Variance: σ² = λ

In both distributions, the mean and variance are related. In the Poisson distribution, the mean and variance are equal.
