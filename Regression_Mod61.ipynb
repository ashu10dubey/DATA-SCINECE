
### **Q1. Explain the difference between simple linear regression and multiple linear regression. Provide an example of each.**

- **Simple Linear Regression**: It models the relationship between a dependent variable \( Y \) and a single independent variable \( X \).  
  - Equation:  
  
    $$ Y = \beta_0 + \beta_1X + \varepsilon  $$
  - Example: Predicting house price based on its size (in square feet).

- **Multiple Linear Regression**: It extends simple linear regression by incorporating multiple independent variables.  
  - Equation:  
    $$
    Y = b_0 + b_1X_1 + b_2X_2 + \dots + b_nX_n + \epsilon
    $$  
  - Example: Predicting house price based on size, number of bedrooms, and location.

### **Q2. Discuss the assumptions of linear regression. How can you check whether these assumptions hold in a given dataset?**

1. **Linearity**: The relationship between independent variables and the dependent variable should be linear.
   - **Check**: Scatter plots or residual plots.

2. **Independence**: Observations should be independent.
   - **Check**: Durbin-Watson test for autocorrelation.

3. **Homoscedasticity**: Constant variance of residuals.
   - **Check**: Plot residuals vs. fitted values.

4. **Normality of Residuals**: Residuals should be normally distributed.
   - **Check**: Q-Q plot, Shapiro-Wilk test.

5. **No Multicollinearity**: Independent variables should not be highly correlated.
   - **Check**: VIF (Variance Inflation Factor).

### **Q3. How do you interpret the slope and intercept in a linear regression model? Provide an example using a real-world scenario.**

- **Intercept (\( b_0 \))**: The predicted value of \( Y \) when \( X \) is zero.
- **Slope (\( b_1 \))**: The change in \( Y \) for a one-unit increase in \( X \).

**Example**:  
Predicting salary based on years of experience:  
$$
\text{Salary} = 30,000 + 5,000 \times (\text{Years of Experience})
$$
- **Interpretation**:  
  - Intercept (30,000): A person with 0 years of experience is predicted to earn $30,000.
  - Slope (5,000): Each additional year of experience increases salary by $5,000.

### **Q4. Explain the concept of gradient descent. How is it used in machine learning?**

- **Gradient Descent**: An optimization algorithm used to minimize the cost function in machine learning models.
- **How it Works**:
  - Compute the gradient (partial derivatives).
  - Update parameters (\( w, b \)) iteratively:  
    $$
    w = w - \alpha \frac{\partial J}{\partial w}
    $$
    where \( \alpha \) is the learning rate.
    
  - Stop when convergence is reached.

**Use in ML**:  
Used in linear regression, logistic regression, and deep learning for parameter optimization.

### **Q5. Describe the multiple linear regression model. How does it differ from simple linear regression?**

- **Multiple Linear Regression**: Extends simple linear regression by including multiple predictors.
  - Equation:  
    $$
    Y = b_0 + b_1X_1 + b_2X_2 + \dots + b_nX_n + \epsilon
    $$
  - Difference:
    - **Simple LR**: Only one independent variable.
    - **Multiple LR**: More than one independent variable.

**Example**:  
Predicting house price based on **size, number of bedrooms, and location score**.

---

### **Q6. Explain the concept of multicollinearity in multiple linear regression. How can you detect and address this issue?**

- **Multicollinearity**: When two or more independent variables are highly correlated, leading to unreliable coefficient estimates.

- **Detection**:
  - **Variance Inflation Factor (VIF)**: If VIF > 5, it indicates multicollinearity.
  - **Correlation Matrix**: High correlation (\( > 0.8 \)) between variables.

- **Solutions**:
  - Remove highly correlated variables.
  - Use **Principal Component Analysis (PCA)**.
  - Use **Ridge Regression** (L2 regularization).

### **Q7. Describe the polynomial regression model. How is it different from linear regression?**

- **Polynomial Regression**: A type of regression where the relationship between independent and dependent variables is modeled as an nth-degree polynomial.
  - Equation:  
    $$
    Y = b_0 + b_1X + b_2X^2 + \dots + b_nX^n + \epsilon
    $$

- **Difference from Linear Regression**:
  - **Linear Regression**: Straight-line relationship.
  - **Polynomial Regression**: Can capture curvilinear relationships.

**Example**:  
Predicting house prices where price **increases at an accelerating rate** with size.

---

### **Q8. What are the advantages and disadvantages of polynomial regression compared to linear regression? In what situations would you prefer to use polynomial regression?**

- **Advantages**:
  - Captures **non-linear** relationships.
  - More flexible than simple linear regression.

- **Disadvantages**:
  - **Overfitting** risk if degree is too high.
  - Computationally expensive.
  - Harder to interpret compared to linear models.

- **When to Use**:
  - When the relationship between variables is **non-linear** (e.g., growth curves, temperature variations).
