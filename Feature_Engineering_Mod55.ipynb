**Q1. What is Min-Max scaling, and how is it used in data preprocessing? Provide an example to illustrate its application.**

Min-Max scaling (also known as normalization) is a feature scaling technique that transforms data to a fixed range, typically [0, 1] or [-1, 1]. It's useful when features have different scales, preventing features with larger values from dominating the model.  It's important to note that Min-Max scaling is sensitive to outliers. If your data contains outliers, StandardScaler (z-score normalization) might be a better choice.  Also, some algorithms, like neural networks, often perform better when features are on a similar scale.

**Formula:**

```
X' = (X - X_min) / (X_max - X_min) * (new_max - new_min) + new_min
```

Where:

*   `X'` is the scaled value.
*   `X` is the original value.
*   `X_min` is the minimum value of the feature.
*   `X_max` is the maximum value of the feature.
*   `new_min` is the desired minimum value of the range (e.g., 0 or -1).
*   `new_max` is the desired maximum value of the range (e.g., 1).

**Example:**

Suppose we have exam scores: [20, 40, 60, 80, 100]

Scaling to the range [0, 1]:

*   `X_min = 20`
*   `X_max = 100`

For the value 40:

```
X' = (40 - 20) / (100 - 20) * (1 - 0) + 0 = 0.25
```

Scaled values: [0.0, 0.25, 0.50, 0.75, 1.0]

**Q2. What is the Unit Vector technique in feature scaling, and how does it differ from Min-Max scaling? Provide an example to illustrate its application.**

The Unit Vector technique (also known as Normalization using L2 norm) scales each data point so that it has a unit norm (magnitude = 1). It's primarily used when the *direction* of the data is more important than its *magnitude*, such as in text processing and cosine similarity calculations (which are based on angles).

**Formula:**

```
X' = X / ||X||
```

Where:

*   `X'` is the scaled vector.
*   `X` is the original vector.
*   `||X||` is the Euclidean norm (L2 norm) of the vector:  `sqrt(x1^2 + x2^2 + ... + xn^2)`

**Difference from Min-Max Scaling:**

*   Min-Max Scaling transforms data to a specific range.
*   Unit Vector Scaling scales data to have a unit length, preserving the direction of the data.

**Example:**

For a data point X = [3, 4]:

```
||X|| = sqrt(3^2 + 4^2) = sqrt(9 + 16) = 5
```

Normalized vector:

```
X' = [3/5, 4/5] = [0.6, 0.8]
```

**Q3. What is PCA (Principal Component Analysis), and how is it used in dimensionality reduction? Provide an example to illustrate its application.**

Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms correlated features into a smaller set of uncorrelated features called Principal Components.  It retains the most important information (variance) in the data while reducing its dimensionality.

**Steps in PCA:**

1.  **Standardize the data (mean=0, std=1):** This is crucial because PCA is sensitive to the scales of the features.  Standardization ensures that all features have equal importance.
2.  **Compute the covariance matrix:** The covariance matrix shows the relationships between features.  It tells us how features vary together.
3.  **Compute eigenvalues & eigenvectors:** Eigenvectors represent the directions of maximum variance in the data, and eigenvalues represent the magnitude of variance along those directions.
4.  **Select the top *k* eigenvectors:** Sort the eigenvalues in descending order. The eigenvectors corresponding to the largest *k* eigenvalues are the principal components that capture the most variance.
5.  **Transform the original dataset:** Project the original data onto the subspace formed by the selected principal components.

**Example:**

Imagine a dataset with height, weight, and age.  PCA might find that height and weight are highly correlated.  It then creates PC1, which captures most of the variance in height and weight, and PC2, which might be more related to age.  Instead of using all three features, we can use just the top two PCs.

**Q4. What is the relationship between PCA and Feature Extraction, and how can PCA be used for Feature Extraction? Provide an example to illustrate this concept.**

Feature Extraction aims to derive new, more informative features from existing ones. PCA is a feature extraction technique because it creates new features (the principal components) that are linear combinations of the original features. These new features capture the most important variance in the data.

**How PCA is used for Feature Extraction:**

1.  PCA transforms the original features into a new set of uncorrelated features (principal components).
2.  Each principal component is a linear combination of the original features.
3.  We retain only the top *k* principal components, effectively reducing the number of features while preserving as much information as possible.

**Example:**

Consider a dataset with 10 sensor readings. PCA might find that 3 principal components capture 95% of the variance. We can then use these 3 principal components as our new features, reducing the complexity of our model.

**Q6. You are working on a project to build a model to predict stock prices. The dataset contains many features, such as company financial data and market trends. Explain how you would use PCA to reduce the dimensionality of the dataset.**

1.  Standardize the dataset (mean=0, std=1) as PCA is sensitive to feature scaling.
2.  Compute the covariance matrix of the features.
3.  Find eigenvalues & eigenvectors and sort them in descending order based on the eigenvalues.
4.  Select the top-k Principal Components that capture a sufficiently high percentage of the variance (e.g., >95%).  This is determined by looking at the *cumulative* explained variance ratio.
5.  Transform the dataset into the new subspace defined by the selected principal components.

By reducing dimensions, PCA helps improve model efficiency and prevents overfitting.  However, it's important to acknowledge that there's a potential loss of information. The goal is to minimize this loss while achieving significant dimensionality reduction.

**Q8. For a dataset containing the following features: [height, weight, age, gender, blood pressure], perform Feature Extraction using PCA. How many principal components would you choose to retain, and why?**

1.  **Preprocessing:**
    *   **Numerical Features:** Standardize the numerical features (height, weight, age, blood pressure) to have zero mean and unit variance. PCA is sensitive to feature scaling, so this step is essential.
    *   **Categorical Feature (Gender):**  PCA works with numerical data. Convert the categorical feature "gender" into numerical representation using one-hot encoding. This creates two new binary features (e.g., "gender_male" and "gender_female").

2.  **PCA:**
    *   Apply PCA to the *numerical* features (standardized height, weight, age, blood pressure, and the one-hot encoded gender features).
    *   Compute the covariance matrix of these features.
    *   Calculate the eigenvalues and eigenvectors of the covariance matrix.
    *   Sort the eigenvalues in descending order.

3.  **Component Selection:**
    *   Calculate the *explained variance ratio* for each principal component. This represents the proportion of the total variance in the data explained by that component.
    *   Calculate the *cumulative explained variance*. This is the sum of the explained variance ratios of the top *k* components.
    *   Choose the number of principal components (*k*) such that the cumulative explained variance reaches a desired threshold (e.g., 95% or 99%).  This means the selected components capture a large proportion of the original data's variance.

4.  **Example:**

| **Principal Component** | **Explained Variance Ratio** | **Cumulative Explained Variance** |
|------------------------|--------------------------|-------------------------------|
| **PC1** | 0.60 | 0.60 |
| **PC2** | 0.25 | 0.85 |
| **PC3** | 0.10 | 0.95 |
| **PC4** | 0.04 | 0.99 |
| **PC5** | 0.01 | 1.00 | |

This table clearly represents the explained variance ratios and cumulative explained variance after performing PCA.
    In this example, if you aim for 95% explained variance, you would retain the first three principal components (PC1, PC2, and PC3). These three components capture 95% of the variance in the original five features. You have effectively reduced the dimensionality from 5 to 3 while preserving most of the information.

**Why this approach?**

*   Standardizing the numerical features ensures that features with larger scales don't dominate PCA.
*   One-hot encoding handles the categorical variable "gender" appropriately for PCA.
*   Using the explained variance ratio and cumulative explained variance helps you choose the optimal number of principal components, balancing dimensionality reduction with information preservation.  You avoid arbitrarily choosing the number of components and ensure you're capturing a sufficient amount of the data's variability.
