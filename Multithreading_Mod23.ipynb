### **Q1: What is multithreading in Python? Why is it used? Name the module used to handle threads in Python.**  
**Answer:**  
Multithreading in Python is a technique that allows multiple threads to run concurrently within a single process. It is used to improve performance by enabling parallel execution of tasks, especially in I/O-bound operations. However, due to Python’s Global Interpreter Lock (GIL), multithreading does not provide true parallelism for CPU-bound tasks.  

The module used to handle threads in Python is the **`threading`** module.  

---

### **Q2: Why is the threading module used? Write the use of the following functions.**  
**Answer:**  
The `threading` module is used to create and manage multiple threads efficiently in Python programs. It helps in handling concurrency and allows programs to run multiple tasks simultaneously.  

**Functions:**  
- **`activeCount()`**: Returns the number of active threads currently running.  
- **`currentThread()`**: Returns the currently executing thread object.  
- **`enumerate()`**: Returns a list of all active threads.

### **Q3: Explain the following functions.**  
**Answer:**  

- **`run()`**: This method contains the code that the thread should execute. It is called internally when a thread is started using `start()`.  
- **`start()`**: Begins the execution of a thread by calling the `run()` method.  
- **`join()`**: Waits for the thread to finish execution before continuing the main program.  
- **`isAlive()`**: Returns `True` if the thread is still running; otherwise, it returns `False`.

### **Q4: Write a Python program to create two threads. Thread one must print the list of squares, and thread two must print the list of cubes.**  

```python
import threading

def print_squares():
    for i in range(1, 6):
        print(f"Square of {i}: {i ** 2}")

def print_cubes():
    for i in range(1, 6):
        print(f"Cube of {i}: {i ** 3}")

# Creating two threads
thread1 = threading.Thread(target=print_squares)
thread2 = threading.Thread(target=print_cubes)

# Starting the threads
thread1.start()
thread2.start()

# Waiting for threads to complete
thread1.join()
thread2.join()

print("Threads execution completed.")
```

### **Q5: State advantages and disadvantages of multithreading.**  
**Answer:**  

#### **Advantages of Multithreading:**  
- **Increased efficiency**: Helps in performing multiple tasks concurrently.  
- **Improved responsiveness**: Useful in GUI applications where tasks like file loading or background processing run without freezing the interface.  
- **Resource sharing**: Threads share the same memory space, reducing memory usage.  

#### **Disadvantages of Multithreading:**  
- **GIL limitation**: In Python, due to the Global Interpreter Lock, true parallel execution is not achieved for CPU-bound tasks.  
- **Complex debugging**: Debugging multithreaded programs can be difficult due to race conditions and deadlocks.  
- **Synchronization overhead**: Managing shared resources requires locks, which can slow down performance.

### **Q6: Explain deadlocks and race conditions.**  
**Answer:**  

- **Deadlock**: A situation where two or more threads are waiting for each other to release resources, causing an infinite waiting loop.  
  - **Example:** If Thread A locks Resource X and needs Resource Y (locked by Thread B), and Thread B locks Resource Y and needs Resource X, both threads get stuck waiting indefinitely.  

- **Race Condition**: A condition that occurs when multiple threads access and modify shared resources concurrently, leading to unpredictable results.  
  - **Example:** If two threads try to increment a shared variable simultaneously, the final value may be incorrect due to uncoordinated access.  

These issues can be mitigated using synchronization mechanisms like **locks (`threading.Lock`)**, **semaphores**, and **thread-safe data structures**.
