**Q1. What are Estimation Statistics? Explain point estimate and interval estimate.**

**A1.**

* **Estimation Statistics:** This branch of statistics deals with using sample data to make inferences about the characteristics of a larger population.

* **Point Estimate:** A single value that is used to estimate a population parameter. For example, the sample mean is a point estimate of the population mean.

* **Interval Estimate:** A range of values within which the population parameter is likely to fall. This range is typically expressed as a confidence interval. For example, a 95% confidence interval for the population mean indicates that we are 95% confident that the true population mean lies within that interval.

* **Why is it used?**
    - To determine if there is enough evidence to support a claim about a population.
    - To make informed decisions in various fields, such as medicine, finance, and social sciences.

* **Importance of Hypothesis Testing:**
    - Helps to make data-driven decisions.
    - Enables researchers to draw conclusions about populations based on sample data.
    - Provides a framework for evaluating the strength of evidence.

**Q5. Write a Python script to conduct a hypothesis test on the difference between two population means, given a sample from each population.**

```python
import scipy.stats as stats

def two_sample_t_test(sample1, sample2, alpha=0.05):
  """
  Conducts a two-sample t-test to compare the means of two populations.

  Args:
    sample1: The first sample.
    sample2: The second sample.
    alpha: Significance level.

  Returns:
    t-statistic, p-value
  """
  t_stat, p_value = stats.ttest_ind(sample1, sample2)
  return t_stat, p_value

# Example Usage:
sample1 = [70, 75, 80, 72, 78]
sample2 = [65, 68, 70, 62, 67]

t_stat, p_value = two_sample_t_test(sample1, sample2)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
  print("Reject the null hypothesis. There is evidence of a significant difference between the means.")
else:
  print("Fail to reject the null hypothesis. There is no evidence of a significant difference between the means.")
```

**Q10. Write a Python program to calculate the two-sample t-test for independent samples, given two
random samples of equal size and a null hypothesis that the population means are equal.**

```python
import scipy.stats as stats

def two_sample_t_test(sample1, sample2, alpha=0.05):
  """
  Conducts a two-sample t-test to compare the means of two populations.

  Args:
    sample1: The first sample.
    sample2: The second sample.
    alpha: Significance level.

  Returns:
    t-statistic, p-value
  """
  t_stat, p_value = stats.ttest_ind(sample1, sample2)
  return t_stat, p_value

# Example Usage:
sample1 = [70, 75, 80, 72, 78]
sample2 = [65, 68, 70, 62, 67]

t_stat, p_value = two_sample_t_test(sample1, sample2)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
  print("Reject the null hypothesis. There is evidence of a significant difference between the means.")
else:
  print("Fail to reject the null hypothesis. There is no evidence of a significant difference between the means.")
```

**Q12. What is t-statistic? State the formula for t-statistic.**

**A12.**

* **t-statistic:** A measure of how far a sample mean is from the population mean in units of standard error.

* **Formula for t-statistic (one-sample t-test):**

   ```
   t = (sample_mean - population_mean) / (sample_std / sqrt(sample_size))
   ```

* **Formula for t-statistic (two-sample t-test):**

   ```
   t = (sample1_mean - sample2_mean) / sqrt((sample1_std^2 / sample1_size) + (sample2_std^2 / sample2_size))
   ```

**Q13. A coffee shop owner wants to estimate the average daily revenue for their shop. They take a random sample of 50 days and find the sample mean revenue to be $500 with a standard deviation of $50. Estimate the population mean revenue with a 95% confidence interval.**

**A13.**

**Assumptions:**

* The sample is a simple random sample.
* The population of daily revenues is normally distributed (or the sample size is large enough for the Central Limit Theorem to apply).

**Formula:**

Confidence Interval = Sample Mean ± (Critical Value) * (Standard Error)

* Standard Error (SE) = Standard Deviation / sqrt(Sample Size)

**Calculations:**

1. **Standard Error (SE):**
   SE = 50 / sqrt(50) = 7.07

2. **Critical Value:**
   For a 95% confidence interval, the critical value (z-score) from the standard normal distribution is 1.96.

3. **Confidence Interval:**
   Lower Bound = 500 - (1.96 * 7.07) = 486.04
   Upper Bound = 500 + (1.96 * 7.07) = 513.96

**Conclusion:**

We are 95% confident that the true population mean daily revenue for the coffee shop lies between $486.04 and $513.96.

**Q14. A researcher hypothesizes that a new drug will decrease blood pressure by 10 mmHg. They conduct a clinical trial with 100 patients and find that the sample mean decrease in blood pressure is 8 mmHg with a standard deviation of 3 mmHg. Test the hypothesis with a significance level of 0.05.**

**A14.**

**1. Set up Hypotheses:**

* **Null Hypothesis (H0):** The mean decrease in blood pressure is 10 mmHg or less. (μ <= 10)
* **Alternative Hypothesis (H1):** The mean decrease in blood pressure is greater than 10 mmHg. (μ > 10)

**2. Calculate the t-statistic:**

* t = (Sample Mean - Hypothesized Mean) / (Standard Error)
* Standard Error = Standard Deviation / sqrt(Sample Size) = 3 / sqrt(100) = 0.3
* t = (8 - 10) / 0.3 = -6.67

**3. Determine the critical value:**

* Degrees of freedom (df) = n - 1 = 100 - 1 = 99
* Using a t-distribution table or a calculator, find the critical value for a one-tailed test with df = 99 and alpha = 0.05. The critical value is approximately 1.66.

**4. Compare t-statistic to critical value:**

* The calculated t-statistic (-6.67) is less than the critical value (1.66).

**5. Decision:**

* Since the t-statistic falls in the rejection region, we reject the null hypothesis.

**6. Conclusion:**

* There is sufficient evidence to support the claim that the new drug decreases blood pressure by more than 10 mmHg at a significance level of 0.05.

**Q17. A marketing company wants to estimate the average number of ads watched by viewers during a TV program. They take a random sample of 50 viewers and find that the sample mean is 4 with a standard deviation of 1.5. Estimate the population mean with a 99% confidence interval.**

**A17.**

**Assumptions:**

* The sample is a simple random sample.
* The population of the number of ads watched is normally distributed (or the sample size is large enough for the Central Limit Theorem to apply).

**Formula:**

Confidence Interval = Sample Mean ± (Critical Value) * (Standard Error)

* Standard Error (SE) = Standard Deviation / sqrt(Sample Size)

**Calculations:**

1. **Standard Error (SE):**
   SE = 1.5 / sqrt(50) = 0.212

2. **Critical Value:**
   For a 99% confidence interval, the critical value (z-score) from the standard normal distribution is 2.576.

3. **Confidence Interval:**
   Lower Bound = 4 - (2.576 * 0.212) = 3.46
   Upper Bound = 4 + (2.576 * 0.212) = 4.54

**Conclusion:**

We are 99% confident that the true population mean number of ads watched by viewers during the TV program lies between 3.46 and 4.54.
