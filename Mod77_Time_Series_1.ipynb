### **Q1. What is a time series, and what are some common applications of time series analysis?**  
A time series is a sequence of data points collected at successive time intervals. It is used to track trends, patterns, and dependencies over time.  
**Applications include:**  
- **Finance:** Stock price prediction  
- **Retail:** Sales forecasting  
- **Weather Forecasting:** Temperature and rainfall prediction  
- **Healthcare:** Patient monitoring and disease progression analysis  
- **IoT & Sensors:** Predictive maintenance  

---

### **Q2. What are some common time series patterns, and how can they be identified and interpreted?**  
Common time series patterns include:  
- **Trend:** Long-term increase or decrease in values (e.g., rising stock prices)  
- **Seasonality:** Regular and predictable fluctuations (e.g., increased ice cream sales in summer)  
- **Cyclic Patterns:** Repeating patterns but with irregular intervals (e.g., business cycles)  
- **Irregular (Noise):** Random variations with no predictable pattern  

These patterns can be identified using **line plots, moving averages, and decomposition techniques**.  

---

### **Q3. How can time series data be preprocessed before applying analysis techniques?**  
- **Handling Missing Values:** Using interpolation or forward-filling  
- **Smoothing:** Applying moving averages to reduce noise  
- **Transformation:** Log transformation to stabilize variance  
- **Differencing:** To remove trends and make the data stationary  
- **Normalization:** Scaling values for better model convergence  

---

### **Q4. How can time series forecasting be used in business decision-making, and what are some common challenges and limitations?**  
**Use Cases:**  
- **Demand Forecasting:** Helps optimize inventory  
- **Financial Planning:** Predicts future revenue and budgeting  
- **Risk Management:** Detects anomalies in system performance  

**Challenges:**  
- **Data Quality Issues:** Missing or inconsistent data  
- **Non-Stationarity:** Data behavior changing over time  
- **External Factors:** Market conditions and unforeseen events (e.g., COVID-19)  

---

### **Q5. What is ARIMA modeling, and how can it be used to forecast time series data?**  
ARIMA (AutoRegressive Integrated Moving Average) is a statistical model used for forecasting time series data.  
- **AR (p):** Uses past values (lags) to predict future values  
- **I (d):** Differencing to remove trends and make data stationary  
- **MA (q):** Uses past errors to correct predictions  

It is widely used when the data shows trends but not strong seasonality.  

---

### **Q6. How do Autocorrelation Function (ACF) and Partial Autocorrelation Function (PACF) plots help in identifying the order of ARIMA models?**  
- **ACF Plot:** Shows correlation between observations and their past values (used to determine MA order)  
- **PACF Plot:** Shows direct relationships between lags after removing indirect effects (used to determine AR order)  

By analyzing the cutoff points in these plots, we determine appropriate values for **p** and **q** in ARIMA.  

---

### **Q7. What are the assumptions of ARIMA models, and how can they be tested for in practice?**  
**Assumptions:**  
1. **Stationarity:** The data should have a constant mean and variance  
2. **Independence of Residuals:** Residuals should be uncorrelated and normally distributed  

**Testing Methods:**  
- **Augmented Dickey-Fuller (ADF) test:** Checks stationarity  
- **Ljung-Box test:** Checks for autocorrelation in residuals  
- **Q-Q Plot:** Tests for normality of residuals

### **Q8. Suppose you have monthly sales data for a retail store for the past three years. Which type of time series model would you recommend for forecasting future sales, and why?**  
I would recommend using **SARIMA (Seasonal ARIMA)** because:  
- It extends ARIMA to handle **seasonal patterns**  
- Monthly sales data typically exhibits **seasonality and trends**  
- SARIMA considers both **seasonal (P, D, Q, s) and non-seasonal (p, d, q) components**  

If the dataset is large, **LSTM (Long Short-Term Memory) neural networks** can also be explored.  

---

### **Q9. What are some of the limitations of time series analysis? Provide an example of a scenario where the limitations of time series analysis may be particularly relevant.**  
**Limitations:**  
- **Assumes Past Trends Continue:** Cannot predict structural changes  
- **Sensitive to Outliers:** Unexpected events (e.g., COVID-19) can disrupt predictions  
- **Data Quality Dependency:** Missing or irregular data affects accuracy  

**Example:**  
In the **stock market**, time series models alone may not work well due to **external economic factors, political events, or company-specific news** that influence prices unpredictably.  

---

### **Q10. Explain the difference between a stationary and non-stationary time series. How does the stationarity of a time series affect the choice of forecasting model?**  
- **Stationary Series:** Mean and variance remain constant over time (e.g., white noise)  
- **Non-Stationary Series:** Mean or variance change over time (e.g., stock prices with an increasing trend)  

**Impact on Forecasting Models:**  
- ARIMA requires **stationary data**, so differencing is applied if needed  
- If the data has **seasonality**, SARIMA or deep learning models (LSTM) may be preferred
