Q5. What are Custom Exceptions in python? Why do we need Custom Exceptions? Explain with an example.

Custom exceptions are user-defined exceptions for specific errors.
 They help in making error messages more meaningful.

```python
class NegativeNumberError(Exception):  # Custom Exception
    pass

def check_positive(num):
    if num < 0:
        raise NegativeNumberError("Number cannot be negative!")
    print("Valid number")

try:
    check_positive(-5)
except NegativeNumberError as e:
    print("Exception:", e)
```

Q6. Create a custom exception class. Use this class to handle an exception.

```python
class AgeTooSmallError(Exception):
    def __init__(self, message="Age is too small!"):
        self.message = message
        super().__init__(self.message)

def validate_age(age):
    if age < 18:
        raise AgeTooSmallError
    print("Valid age")

try:
    validate_age(16)
except AgeTooSmallError as e:
    print("Exception:", e)
```
