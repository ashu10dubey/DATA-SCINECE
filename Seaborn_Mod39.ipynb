
### **Q1: Name any five plots that we can plot using the Seaborn library. Also, state the uses of each plot.**

#### **Answer:**
Seaborn provides several visualization techniques. Below are five commonly used plots:

1. **Line Plot (`sns.lineplot()`)**  
   - **Use**: Used to display trends over a continuous variable (time series, stock prices, etc.).  
   - **Example**: Showing the trend of sales over months.

2. **Bar Plot (`sns.barplot()`)**  
   - **Use**: Used to compare categorical data with numerical values.  
   - **Example**: Comparing average test scores of different student groups.

3. **Box Plot (`sns.boxplot()`)**  
   - **Use**: Used to visualize the distribution, median, quartiles, and outliers of numerical data.  
   - **Example**: Displaying the distribution of salaries in different departments.

4. **Histogram (`sns.histplot()`)**  
   - **Use**: Used to visualize the frequency distribution of numerical data.  
   - **Example**: Showing the distribution of ages in a population.

5. **Heatmap (`sns.heatmap()`)**  
   - **Use**: Used to visualize relationships in tabular data through color encoding.  
   - **Example**: Displaying a correlation matrix of different features in a dataset.

### **Q2: Load the "fmri" dataset using the load_dataset function of seaborn. Plot a line plot using x = "timepoint" and y = "signal" for different events and regions.**

#### **Solution:**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load fmri dataset
fmri = sns.load_dataset("fmri")

# Plot line plot
plt.figure(figsize=(10, 6))
sns.lineplot(data=fmri, x="timepoint", y="signal", hue="event", style="region")
plt.title("FMRI Signal Over Time for Different Events and Regions")
plt.xlabel("Timepoint")
plt.ylabel("Signal")
plt.legend(title="Event & Region")
plt.show()
```


### **Q3: Load the "titanic" dataset using the load_dataset function of seaborn. Plot two box plots using x = 'pclass', y = 'age' and y = 'fare'.**

#### **Solution:**

```python
# Load titanic dataset
titanic = sns.load_dataset("titanic")

# Create subplots for age and fare
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Boxplot for Age vs Pclass
sns.boxplot(data=titanic, x="pclass", y="age", ax=axes[0])
axes[0].set_title("Age Distribution Across Passenger Classes")

# Boxplot for Fare vs Pclass
sns.boxplot(data=titanic, x="pclass", y="fare", ax=axes[1])
axes[1].set_title("Fare Distribution Across Passenger Classes")

plt.tight_layout()
plt.show()
```


### **Q4: Use the "diamonds" dataset from seaborn to plot a histogram for the 'price' column. Use the hue parameter for the 'cut' column of the diamonds dataset.**

#### **Solution:**

```python
# Load diamonds dataset
diamonds = sns.load_dataset("diamonds")

# Plot histogram
plt.figure(figsize=(10, 6))
sns.histplot(data=diamonds, x="price", hue="cut", kde=True, bins=30)
plt.title("Histogram of Diamond Prices by Cut")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.legend(title="Cut")
plt.show()
```

### **Q5: Use the "iris" dataset from seaborn to plot a pair plot. Use the hue parameter for the "species" column of the iris dataset.**

#### **Solution:**

```python
# Load iris dataset
iris = sns.load_dataset("iris")

# Pairplot
sns.pairplot(iris, hue="species", diag_kind="kde")
plt.suptitle("Pair Plot of Iris Dataset", y=1.02)
plt.show()
```

### **Q6: Use the "flights" dataset from seaborn to plot a heatmap.**

#### **Solution:**

```python
# Load flights dataset
flights = sns.load_dataset("flights")

# Pivot the dataset to create a matrix format for heatmap
# Use keyword arguments for index, columns, and values
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

# Plot heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(flights_pivot, cmap="coolwarm", annot=True, fmt="d")
plt.title("Heatmap of Flight Passengers Over Time")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
```
