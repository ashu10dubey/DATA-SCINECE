### **Q1. Create a Pandas Series that contains the following data: 4, 8, 15, 16, 23, and 42. Then, print the series.**

**Answer:**
A Pandas Series is a one-dimensional labeled array capable of holding any data type. Below is a Python program to create and print a Pandas Series.

```python
import pandas as pd

# Creating a Pandas Series
data = pd.Series([4, 8, 15, 16, 23, 42])

# Printing the Series
print(data)
```

**Output:**
```
0     4
1     8
2    15
3    16
4    23
5    42
dtype: int64
```

### **Q3. Create a Pandas DataFrame that contains the following data and print it.**

| Name   | Age | Gender |
|--------|----|--------|
| Alice  | 25 | Female |
| Bob    | 30 | Male   |
| Claire | 27 | Female |

**Answer:**
```python
import pandas as pd

# Creating a DataFrame
data = {
    "Name": ["Alice", "Bob", "Claire"],
    "Age": [25, 30, 27],
    "Gender": ["Female", "Male", "Female"]
}

df = pd.DataFrame(data)

# Printing the DataFrame
print(df)
```

### **Q4. What is ‘DataFrame’ in pandas and how is it different from pandas.Series? Explain with an example.**

**Answer:**
A **DataFrame** is a two-dimensional data structure in Pandas, similar to an Excel spreadsheet or an SQL table. It consists of multiple columns, where each column is a Pandas Series.

A **Series** is a one-dimensional labeled array that holds data of a single type.

#### **Example:**
```python
import pandas as pd

# Creating a Series
series_example = pd.Series([10, 20, 30, 40])

# Creating a DataFrame
dataframe_example = pd.DataFrame({
    "Numbers": [10, 20, 30, 40],
    "Letters": ["A", "B", "C", "D"]
})

print("Series:")
print(series_example)

print("\nDataFrame:")
print(dataframe_example)
```

**Output:**
```
Series:
0    10
1    20
2    30
3    40
dtype: int64

DataFrame:
   Numbers Letters
0       10       A
1       20       B
2       30       C
3       40       D
```

**Key Difference:**
- A **Series** is a single column of data with labels (like a column in Excel).
- A **DataFrame** is a collection of multiple Series, making it two-dimensional.

### **Q5. What are some common functions you can use to manipulate data in a Pandas DataFrame? Can you give an example of when you might use one of these functions?**

**Answer:**
Some common functions used to manipulate Pandas DataFrames include:

- `df.head(n)`: Displays the first `n` rows.
- `df.tail(n)`: Displays the last `n` rows.
- `df.info()`: Provides information about the DataFrame.
- `df.describe()`: Provides statistical summary.
- `df.drop(columns=['col_name'])`: Drops a column.
- `df.loc[row, column]`: Accesses specific elements.
- `df.groupby('column_name')`: Groups data by a column.

#### **Example: Using `describe()` to get summary statistics**
```python
import pandas as pd

# Creating a DataFrame
data = {
    "Name": ["Alice", "Bob", "Claire", "David"],
    "Age": [25, 30, 27, 35],
    "Salary": [50000, 60000, 55000, 70000]
}

df = pd.DataFrame(data)

# Using describe() to get statistics
print(df.describe())
```

**Output:**
```
             Age       Salary
count   4.00000      4.00000
mean   29.25000  58750.00000
std     4.78714   8539.12564
min    25.00000  50000.00000
25%    26.50000  53750.00000
50%    28.50000  57500.00000
75%    31.25000  62500.00000
max    35.00000  70000.00000
```

### **Q7. Create a DataFrame using multiple Series. Explain with an example.**

**Answer:**
A DataFrame can be created using multiple Series, where each Series represents a column of the DataFrame.

#### **Example:**
```python
import pandas as pd

# Creating multiple Series
names = pd.Series(["Alice", "Bob", "Claire"])
ages = pd.Series([25, 30, 27])
genders = pd.Series(["Female", "Male", "Female"])

# Creating a DataFrame using the Series
df = pd.DataFrame({
    "Name": names,
    "Age": ages,
    "Gender": genders
})

# Printing the DataFrame
print(df)
```

**Output:**
```
    Name  Age  Gender
0  Alice   25  Female
1    Bob   30    Male
2  Claire   27  Female
```

**Explanation:**
- Here, three separate **Series** (`names`, `ages`, and `genders`) are created.
- These Series are combined into a **DataFrame**, where each Series becomes a column.
