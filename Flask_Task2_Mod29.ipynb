
### **Q1. Explain GET and POST methods.**
In Flask, **GET** and **POST** are two common HTTP methods used in web applications:

- **GET Method:**
  - Used to retrieve data from the server.
  - Parameters are sent in the URL as query strings (`?key=value`).
  - Example: `https://example.com/data?id=123`
  - It is idempotent (multiple requests do not change server state).

- **POST Method:**
  - Used to send data to the server (e.g., form submission).
  - Data is sent in the request body.
  - It is not idempotent (each request may modify the database).
  - More secure than GET for sensitive data.

Example Flask code:
```python
from flask import Flask, request

app = Flask(__name__)

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    if request.method == 'POST':
        return "POST request received!"
    return "GET request received!"

if __name__ == '__main__':
    app.run(debug=True)
```

### **Q2. Why is `request` used in Flask?**
- The `request` object in Flask is used to **access incoming request data**.
- It allows retrieving **form data, JSON payload, headers, cookies**, etc.
- Example use cases:
  - Fetching form inputs: `request.form['name']`
  - Getting query parameters: `request.args.get('id')`
  - Receiving JSON data: `request.get_json()`

Example:
```python
from flask import Flask, request

app = Flask(__name__)

@app.route('/data', methods=['GET'])
def get_data():
    user_id = request.args.get('id')  # Get parameter from URL
    return f"User ID: {user_id}"

if __name__ == '__main__':
    app.run(debug=True)
```

### **Q3. Why is `redirect()` used in Flask?**
- The `redirect()` function is used to **redirect users to another route or URL**.
- It is useful for:
  - Redirecting after form submission.
  - Redirecting unauthorized users to a login page.

Example:
```python
from flask import Flask, redirect, url_for

app = Flask(__name__)

@app.route('/')
def home():
    return "Welcome to Home Page!"

@app.route('/login')
def login():
    return redirect(url_for('home'))  # Redirect to home page

if __name__ == '__main__':
    app.run(debug=True)
```

### **Q4. What are templates in Flask? Why is `render_template()` used?**
- **Templates** in Flask are HTML files with dynamic content placeholders.
- They allow separating business logic from presentation logic.
- The **`render_template()`** function is used to render HTML templates dynamically.
- Flask uses **Jinja2** templating engine (`{{ variable }}` for placeholders).

Example:
```python
from flask import Flask, render_template

app = Flask(__name__)

@app.route('/')
def home():
    return render_template('index.html', name="Ashutosh")  # Pass variable

if __name__ == '__main__':
    app.run(debug=True)
```

Example `index.html` (inside the `templates` folder):
```html
<!DOCTYPE html>
<html>
<head>
    <title>Flask Template</title>
</head>
<body>
    <h1>Hello, {{ name }}!</h1>
</body>
</html>
```

### **Q5. Create a simple API and test it in Postman**
#### **Step 1: Create Flask API**
```python
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/api/data', methods=['GET'])
def get_data():
    data = {"message": "Hello from Flask API!", "status": "success"}
    return jsonify(data)

if __name__ == '__main__':
    app.run(debug=True)
```

#### **Step 2: Test in Postman**
1. Open **Postman**.
2. Send a **GET request** to `http://127.0.0.1:5000/api/data`.
3. Response will be:
   ```json
   {
       "message": "Hello from Flask API!",
       "status": "success"
   }
   ```
