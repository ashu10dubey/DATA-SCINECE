Q1. Which function is used to open a file? What are the different modes of opening a file? Explain each mode of file opening.

````python
'''The open() function is used to open a file in Python.
Different modes of opening a file:
'r' (Read mode): Opens the file for reading. If the file does not exist, it raises an error.
'w' (Write mode): Opens the file for writing. If the file exists, it truncates (erases) its content. If the file does not exist, it creates a new one.
'a' (Append mode): Opens the file for writing, but does not erase existing content. It adds data to the end of the file.
'x' (Exclusive creation mode): Creates a new file but raises an error if the file already exists.
'b' (Binary mode): Used with other modes (rb, wb, ab, etc.) for reading/writing binary files like images or videos.
't' (Text mode): Default mode, used for reading and writing text files.
'+' (Read and Write mode): Allows both reading and writing. Used as 'r+', 'w+', 'a+'.'''
````

Q2. Why close() function is used? Why is it important to close a file?

The close() function is used to close an opened file.
Importance of closing a file:
Frees up system resources.
Ensures all data is properly written to the file (especially when writing).
Prevents file corruption or unexpected behavior.

````python
file = open("example.txt", "w")
file.write("Hello, World!")
file.close()  # Closing the file
````

Q3. Write a python program to create a text file. Write ‘I want to become a Data Scientist’ in that file. Then
close the file. Open this file and read the content of the file.

````python
# Creating and writing to the file
file = open("data_scientist.txt", "w")
file.write("I want to become a Data Scientist")
file.close()

# Opening and reading the file
file = open("data_scientist.txt", "r")
content = file.read()
print(content)
file.close()
````

Q4. Explain the following with python code:
read(): Reads the entire file as a string.
readline(): Reads one line at a time.
readlines(): Reads all lines into a list.


````python
# Writing sample data
with open("sample.txt", "w") as file:
    file.write("First line\nSecond line\nThird line")

# Using read()
with open("sample.txt", "r") as file:
    print("Using read():")
    print(file.read())

# Using readline()
with open("sample.txt", "r") as file:
    print("\nUsing readline():")
    print(file.readline())  # Reads first line
    print(file.readline())  # Reads second line

# Using readlines()
with open("sample.txt", "r") as file:
    print("\nUsing readlines():")
    print(file.readlines())  # Reads all lines as a list
````

Q5. Explain why with statement is used with open(). What is the advantage of using with statement and
open() together?

The with statement is used to handle file operations safely.
Advantages:
Automatically closes the file after the block execution.
Prevents resource leakage.
No need to manually call close().

````python
with open("example.txt", "w") as file:
    file.write("This is a test file.")

# No need to call file.close(), it is automatically closed
````

Q6. Explain the write() and writelines() functions. Give a suitable example.

````python
# Using write()
with open("write_example.txt", "w") as file:
    file.write("Hello, World!\n")
    file.write("Python File Handling")

# Using writelines()
lines = ["Line 1\n", "Line 2\n", "Line 3\n"]
with open("writelines_example.txt", "w") as file:
    file.writelines(lines)
````
