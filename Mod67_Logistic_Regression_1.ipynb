### **Q1. Explain the difference between linear regression and logistic regression models. Provide an example of a scenario where logistic regression would be more appropriate.**  

| Feature              | **Linear Regression**                                    | **Logistic Regression**                              |
|----------------------|----------------------------------------------------------|------------------------------------------------------|
| **Type of problem**  | Predicts continuous values                               | Predicts probabilities (classification)             |
| **Equation**        | $$ y = wX + b $$                                     | $$ P(y=1) = \frac{1}{1 + e^{-(wX + b)}} $$ |
| **Output**          | Real numbers                                             | Values between 0 and 1 (probabilities)              |
| **Use case**        | House price prediction, sales forecasting                 | Spam detection, fraud detection                      |

**Example:**  
Logistic regression is more appropriate for **email classification (spam or not spam)** since the outcome is binary.

### **Q2. What is the cost function used in logistic regression, and how is it optimized?**  
Logistic regression uses the **log loss (binary cross-entropy) cost function**:

$$ J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right] $$

where:  
- $$ y_i $$= actual label (0 or 1)  
- $$ \hat{y}_i $$ = predicted probability
- \( m \) = number of samples  

**Optimization:**  
- Optimized using **Gradient Descent**, where the parameters are updated as:  

  $$ \theta := \theta - \alpha \frac{\partial J(\theta)}{\partial \theta} $$  

  where \( \alpha \) is the learning rate.

### **Q3. Explain the concept of regularization in logistic regression and how it helps prevent overfitting.**  
Regularization **penalizes large coefficients** to prevent overfitting.  

- **L1 Regularization (Lasso):**  

  $$ J(\theta) = -\frac{1}{m} \sum \left[ y \log \hat{y} + (1 - y) \log (1 - \hat{y}) \right] + \lambda \sum |\theta_j| $$  

  - Shrinks some coefficients to **zero**, leading to **feature selection**.

- **L2 Regularization (Ridge):**  

  $$ J(\theta) = -\frac{1}{m} \sum \left[ y \log \hat{y} + (1 - y) \log (1 - \hat{y}) \right] + \lambda \sum \theta_j^2 $$  

  - Reduces coefficient size but **does not remove features**.

- **Elastic Net (Combination of L1 & L2):**  

  $$ J(\theta) = -\frac{1}{m} \sum \left[ y \log \hat{y} + (1 - y) \log (1 - \hat{y}) \right] + \lambda_1 \sum |\theta_j| + \lambda_2 \sum \theta_j^2 $$  

Regularization helps in cases where:  
- The model is **overfitting** due to too many features.  
- There is **multicollinearity** (L2 helps stabilize feature importance).

### **Q4. What is the ROC curve, and how is it used to evaluate the performance of the logistic regression model?**  
The **ROC (Receiver Operating Characteristic) curve** plots:  
- **True Positive Rate (Recall)**:  

  $$ TPR = \frac{TP}{TP + FN} $$  

- **False Positive Rate (1 - Specificity)**:  

  $$ FPR = \frac{FP}{FP + TN} $$  

**Area Under the Curve (AUC-ROC Score)**:  
- If **AUC = 1** → **Perfect classifier**  
- If **AUC = 0.5** → **Random guessing**  
- If **AUC < 0.5** → **Worse than random**  

**Usage:**  
- Helps choose an optimal **threshold** for classification.  
- A higher AUC means a **better-performing** model.

### **Q5. What are some common techniques for feature selection in logistic regression? How do these techniques help improve the model's performance?**  
1. **Recursive Feature Elimination (RFE):** Iteratively removes least important features.  
2. **Lasso Regression (L1 Regularization):** Shrinks some coefficients to **zero**, effectively selecting important features.  
3. **Variance Threshold:** Removes low-variance features that do not contribute much.  
4. **Mutual Information:** Measures dependency between features and target.  
5. **Chi-Square Test:** Used for categorical features.  

**Benefits:**  
- Reduces **overfitting**.  
- Speeds up training.  
- Improves **interpretability**.

### **Q6. How can you handle imbalanced datasets in logistic regression? What are some strategies for dealing with class imbalance?**  
**Strategies:**  
1. **Resampling:**  
   - **Oversampling** minority class (e.g., **SMOTE**).  
   - **Undersampling** majority class.  

2. **Class Weight Adjustment:**  
   - Adjusts the penalty for misclassified minority samples.  
   - Use `class_weight='balanced'` in `sklearn`.  

3. **Synthetic Data Generation:**  
   - **SMOTE (Synthetic Minority Over-sampling Technique)** creates synthetic samples.  

4. **Threshold Moving:**  
   - Adjusting the classification threshold based on precision-recall tradeoff.  

5. **Use Alternative Metrics:**  
   - **Precision, Recall, F1-score, AUC-ROC** instead of accuracy.

### **Q7. Can you discuss some common issues and challenges that may arise when implementing logistic regression, and how they can be addressed?**  

| **Issue**               | **Solution** |
|-------------------------|-------------|
| **Multicollinearity** (high correlation between independent variables) | Use **VIF (Variance Inflation Factor)**, apply **L2 Regularization (Ridge Regression)** |
| **Imbalanced Classes** (skewed label distribution) | Use **oversampling, undersampling, or SMOTE** |
| **Outliers** (affect logistic regression coefficients) | **Remove or transform outliers** using scaling techniques |
| **Feature Scaling Required** | Apply **Standardization (Z-score) or MinMax Scaling** |
| **Overfitting** (too many features) | Use **Regularization (L1, L2) or Feature Selection** |
| **Convergence Issues** (Gradient Descent fails) | Reduce **learning rate**, normalize data, or use different solver (e.g., `lbfgs`, `saga`) |
