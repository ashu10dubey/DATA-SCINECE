Q1. **What is the role of feature selection in anomaly detection?**  

Feature selection helps improve anomaly detection by:  
- **Reducing Noise:** Removing irrelevant or redundant features enhances model performance.  
- **Improving Interpretability:** Helps in understanding which features contribute to anomalies.  
- **Enhancing Efficiency:** Reduces computational complexity, especially in high-dimensional datasets.  
- **Boosting Detection Accuracy:** Selecting the right features improves separation between normal and anomalous data.  

Mathematically, feature selection can be performed using **Mutual Information (MI)**:  
$$ MI(X, Y) = \sum_{x \in X} \sum_{y \in Y} P(x, y) \log \left(\frac{P(x, y)}{P(x)P(y)}\right) $$  
where higher MI values indicate more relevant features.

Q2. **What are some common evaluation metrics for anomaly detection algorithms and how are they computed?**  

1. **Precision**: Measures how many detected anomalies are actual anomalies.  
   $$ \text{Precision} = \frac{TP}{TP + FP} $$  
2. **Recall (Sensitivity)**: Measures how many actual anomalies were detected.  
   $$ \text{Recall} = \frac{TP}{TP + FN} $$  
3. **F1-Score**: Harmonic mean of precision and recall.  
   $$ F1 = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} $$  
4. **ROC-AUC Score**: Measures the area under the Receiver Operating Characteristic curve.  
5. **Adjusted Rand Index (ARI)**: Used in clustering-based anomaly detection.  
6. **Average Path Length** (Isolation Forest):  
   $$ S(x) = 2^{-\frac{E(h(x))}{c(n)}} $$  
   where shorter path length indicates higher anomaly.

Q3. **What is DBSCAN and how does it work for clustering?**  

DBSCAN (**Density-Based Spatial Clustering of Applications with Noise**) groups points based on density. It works by:  
1. Identifying **core points** (having at least \( minPts \) neighbors within radius \( \epsilon \)).  
2. Expanding clusters from core points.  
3. Labeling points as **border** (connected to core points) or **noise** (outliers).  

DBSCAN does not require specifying the number of clusters, making it useful for anomaly detection.  

---

Q4. **How does the epsilon parameter affect the performance of DBSCAN in detecting anomalies?**  

- **Small \( \epsilon \):** More points are labeled as noise (increases false positives).  
- **Large \( \epsilon \):** Fewer anomalies detected as more points are clustered together (increases false negatives).  
- **Optimal \( \epsilon \)** can be found using the **k-distance plot**, where the "elbow point" suggests a good choice.  

---

Q5. **What are the differences between the core, border, and noise points in DBSCAN, and how do they relate to anomaly detection?**  

1. **Core Points:** Have at least \( minPts \) neighbors within \( \epsilon \). They form cluster centers.  
2. **Border Points:** Have fewer than \( minPts \) neighbors but are close to core points.  
3. **Noise Points:** Do not belong to any cluster and are considered anomalies.  

Anomalies in DBSCAN are typically **noise points** that do not fit into any dense cluster.  

---

Q6. **How does DBSCAN detect anomalies and what are the key parameters involved in the process?**  

DBSCAN detects anomalies as **noise points**, which are isolated and not part of any dense region. The key parameters are:  
- **\( \epsilon \) (Epsilon):** Defines the neighborhood radius.  
- **\( minPts \):** Minimum number of points required to form a dense region.  

A point is an **anomaly** if it has fewer than \( minPts \) neighbors within \( \epsilon \).
