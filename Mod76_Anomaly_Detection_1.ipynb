Q1. **What is anomaly detection and what is its purpose?**  

Anomaly detection is the process of identifying rare or unusual patterns in data that do not conform to expected behavior. It is used in applications such as fraud detection, network security, predictive maintenance, and fault detection.  

---

Q2. **What are the key challenges in anomaly detection?**  

Some major challenges in anomaly detection include:  
- **High Dimensionality:** Traditional distance-based methods fail in high-dimensional spaces.  
- **Imbalanced Data:** Anomalies are rare, making it difficult for models to learn effectively.  
- **Concept Drift:** The definition of anomalies may change over time.  
- **Lack of Labeled Data:** Many real-world datasets lack labeled anomalies, making supervised learning difficult.  

---

Q3. **How does unsupervised anomaly detection differ from supervised anomaly detection?**  

- **Supervised Anomaly Detection:** Requires labeled data (normal vs. anomalous). It is trained like a classification problem.  
- **Unsupervised Anomaly Detection:** Works without labeled data. It assumes that anomalies are rare and different from the majority of the data.  

---

Q4. **What are the main categories of anomaly detection algorithms?**  

1. **Statistical Methods:** Assume anomalies deviate significantly from a normal distribution. Example: Z-score, Gaussian Mixture Models (GMM).  
2. **Distance-Based Methods:** Use distance measures to find outliers. Example: K-Nearest Neighbors (KNN), LOF.  
3. **Clustering-Based Methods:** Assume anomalies do not belong to clusters. Example: DBSCAN, K-Means for anomaly detection.  
4. **Machine Learning-Based Methods:** Use models like Isolation Forest and Autoencoders.  
5. **Deep Learning-Based Methods:** Utilize neural networks, such as Variational Autoencoders (VAE) and LSTMs.  

---

Q5. **What are the main assumptions made by distance-based anomaly detection methods?**  

Distance-based methods assume:  
- Anomalies are far from normal data points.  
- Normal data points have dense local neighborhoods, whereas anomalies are isolated.  
- The distance between a point and its nearest neighbors helps determine its anomaly score.  

Mathematically, the anomaly score for a point \( x \) using KNN can be computed as:  
$$ S(x) = \frac{1}{K} \sum_{i=1}^{K} d(x, x_i) $$  
where \( d(x, x_i) \) is the distance between \( x \) and its \( K \) nearest neighbors.  

---

Q6. **How does the LOF algorithm compute anomaly scores?**  

The **Local Outlier Factor (LOF)** algorithm computes anomaly scores based on the **local density** of a point relative to its neighbors.  

1. Compute the **reachability distance** of point \( x \) w.r.t. its \( K \)-th nearest neighbor:  
   $$ \text{reachability-dist}(x, y) = \max(\text{k-distance}(y), d(x, y)) $$  

2. Compute the **local reachability density (LRD)** for a point \( x \):  
   $$ \text{LRD}(x) = \frac{K}{\sum_{y \in KNN(x)} \text{reachability-dist}(x, y)} $$  

3. Compute the **Local Outlier Factor (LOF)**:  
   $$ \text{LOF}(x) = \frac{\sum_{y \in KNN(x)} \frac{\text{LRD}(y)}{\text{LRD}(x)}}{K} $$  

If \( \text{LOF}(x) > 1 \), then \( x \) is an anomaly. Higher LOF values indicate greater anomaly.

Q7. **What are the key parameters of the Isolation Forest algorithm?**  

1. **Number of Trees (n_estimators)**: Determines the number of isolation trees. More trees improve robustness but increase computation time.  
2. **Subsampling Size (max_samples)**: Number of random samples per tree, affecting sensitivity to anomalies.  
3. **Contamination Parameter**: Controls the proportion of expected anomalies.  
4. **Path Length**: Defines the depth at which a point is isolated; shorter paths indicate anomalies.  

The anomaly score for a point \( x \) in an Isolation Forest is computed as:  
$$ S(x) = 2^{-\frac{E(h(x))}{c(n)}} $$  
where \( E(h(x)) \) is the average path length and \( c(n) \) is the expected path length for a normal point in a dataset of size \( n \).

Q8. **If a data point has only 2 neighbors of the same class within a radius of 0.5, what is its anomaly score using KNN with K=10?**  

The anomaly score using **KNN anomaly detection** is typically computed as:  
$$ S(x) = \frac{1}{K} \sum_{i=1}^{K} d(x, x_i) $$  

- The point has only **2 neighbors within radius 0.5**, meaning **8 of its 10 neighbors** are farther away.  
- If the remaining 8 neighbors are much farther than 0.5, the anomaly score will be **high**.  
- This point is likely to be an anomaly since it is sparsely surrounded by its class.

Q9. **Using the Isolation Forest algorithm with 100 trees and a dataset of 3000 data points, what is the anomaly score for a data point that has an average path length of 5.0 compared to the average path length of the trees?**  

The anomaly score in **Isolation Forest** is given by:  
$$ S(x) = 2^{-\frac{E(h(x))}{c(n)}} $$  

where:  
- \( E(h(x)) = 5.0 \) (observed average path length)  
- \( c(n) \approx 2H(n - 1) - \frac{2(n-1)}{n} \) (expected average path length for normal points)  
- For \( n = 3000 \), \( c(n) \approx 10.5 \)  

Computing the anomaly score:  
$$ S(x) = 2^{-\frac{5.0}{10.5}} $$  
$$ S(x) = 2^{-0.476} \approx 0.72 $$  

Since 0.72 is **relatively high**, the data point is **not likely an anomaly**. Lower scores (e.g., < 0.5) indicate stronger anomalies.
