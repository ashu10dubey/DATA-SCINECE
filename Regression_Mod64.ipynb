### **Q1. What is Lasso Regression, and how does it differ from other regression techniques?**  

#### **Answer:**  
**Lasso Regression (Least Absolute Shrinkage and Selection Operator)** is a type of linear regression that uses **L1 regularization** to constrain the model’s coefficients. It minimizes the sum of squared errors while also adding a penalty proportional to the **absolute** values of the coefficients.

#### **Lasso Regression Formula:**  
$$
\sum (y_i - X\beta)^2 + \lambda \sum |\beta|
$$
where:  
- \( y_i \) is the target variable,  
- \( X \) represents the input features,  
- \( \beta \) are the model coefficients,  
- \( \lambda \) is the regularization parameter.

#### **Differences from Other Regression Techniques:**  
| Feature | OLS Regression | Ridge Regression | Lasso Regression |
|---------|---------------|------------------|------------------|
| Regularization | No | L2 (squared coefficients) | L1 (absolute coefficients) |
| Feature Selection | No | Retains all features | Can shrink some coefficients to exactly zero (automatic selection) |
| Handles Multicollinearity | No | Yes, but keeps all variables | Yes, and removes irrelevant variables |
| Bias-Variance Tradeoff | High variance in multicollinear cases | Lower variance, higher bias | Even higher bias, but better feature selection |

### **Q2. What is the main advantage of using Lasso Regression in feature selection?**  

#### **Answer:**  
The main advantage of **Lasso Regression** is that it **performs automatic feature selection** by shrinking some coefficients **exactly to zero**. This helps:

1. **Reduce model complexity** by eliminating unimportant features.
2. **Improve interpretability** by keeping only the most significant predictors.
3. **Prevent overfitting**, making the model generalize better.

This makes Lasso useful in **high-dimensional datasets** where many features are irrelevant.

---

### **Q3. How do you interpret the coefficients of a Lasso Regression model?**  

#### **Answer:**  
The interpretation of **Lasso Regression coefficients** is similar to ordinary regression, but with key differences:

- **Nonzero coefficients**: Represent the impact of that feature on the target variable.
- **Zero coefficients**: Indicate that the feature has been eliminated from the model.
- **Shrunken coefficients**: Compared to OLS, Lasso pushes some coefficients closer to zero.

#### **Example:**  
A Lasso model predicting house prices gives:
$$
\text{Price} = 50,000 + 30 \times \text{sq_ft} + 0 \times \text{garden_size} + 200 \times \text{num_bedrooms}
$$
- `sq_ft` and `num_bedrooms` contribute to price prediction.
- `garden_size` has been removed (coefficient = 0), meaning it has little impact.

### **Q4. What are the tuning parameters that can be adjusted in Lasso Regression, and how do they affect the model's performance?**  

#### **Answer:**  
The primary tuning parameter in Lasso Regression is **\(\lambda\) (the regularization strength)**.  

- **Small \(lambda\) (near 0):**  
  - Behaves like OLS (minimal regularization).  
  - Coefficients remain large.  
  - Risk of overfitting.  
- **Moderate \(lambda\):**  
  - Balances bias-variance tradeoff.  
  - Shrinks some coefficients but retains important features.  
- **Large \(lambda\):**  
  - Strong regularization.  
  - Many coefficients shrink to zero.  
  - High bias, risk of underfitting.  

#### **Other Hyperparameters:**
- **Normalization (`normalize=True/False`)**: Standardizes input features.
- **Solver choice**: Some solvers are optimized for large datasets.

### **Q5. Can Lasso Regression be used for non-linear regression problems? If yes, how?**  

#### **Answer:**  
Yes, **Lasso Regression can handle non-linear relationships** using **feature engineering**:

1. **Polynomial Features:**  
   - Convert variables into polynomial terms (e.g., \( x^2, x^3 \)).
   - Then apply Lasso to control feature selection.

2. **Interaction Features:**  
   - Include combinations of variables to capture interactions.

3. **Kernel Trick:**  
   - Convert data into a higher-dimensional space (similar to SVMs).

 **Note:** Lasso itself is a linear model, but when combined with polynomial transformations, it can approximate **non-linear relationships**.

### **Q6. What is the difference between Ridge Regression and Lasso Regression?**  

#### **Answer:**  
| Feature | Ridge Regression | Lasso Regression |
|---------|-----------------|------------------|
| Regularization Type | L2 (squared penalty) | L1 (absolute penalty) |
| Effect on Coefficients | Shrinks them but keeps all nonzero | Shrinks some to exactly zero (feature selection) |
| Handles Multicollinearity | Yes | Yes (better for sparse models) |
| Feature Selection | No | Yes |
| Bias-Variance Tradeoff | Lower bias, slightly reduced variance | Higher bias, greater variance reduction |

---

### **Q7. Can Lasso Regression handle multicollinearity in the input features? If yes, how?**  

#### **Answer:**  
Yes, **Lasso Regression handles multicollinearity** by **selecting only one of the correlated variables** and shrinking others to zero.

#### **Example:**
Suppose `Feature A` and `Feature B` are highly correlated.  
- **Ridge Regression** keeps both, but reduces their impact.  
- **Lasso Regression** eliminates one and retains the most important.

This improves **interpretability** by keeping only the most relevant predictors.

### **Q8. How do you choose the optimal value of the regularization parameter (lambda) in Lasso Regression?**  

#### **Answer:**  
Choosing the best **\(lambda\)** is crucial for balancing **bias and variance**.

### **Methods to Select Optimal \(lambda\):**
1. **Cross-Validation:**  
   - Use **k-fold cross-validation** to test different \(\lambda\) values.
   - Choose the \(lambda\) that minimizes prediction error.

2. **Grid Search or Random Search:**  
   - Try multiple \(lambda\) values and find the best.

3. **LassoCV in Python (`sklearn`):**  
   - `LassoCV` automatically selects the best \(lambda\).

4. **L-Curve Method:**  
   - Plots trade-off between error and penalty.

 **Rule of Thumb:**  
- If too many coefficients = 0, **reduce \(lambda)**.  
- If still overfitting, **increase \(lambda)**.
