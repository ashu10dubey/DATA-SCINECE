
### **Q1. What is data encoding? How is it useful in data science?**  
**Answer:**  
Data encoding is the process of converting categorical data into a numerical format so that machine learning models can process it. Since most ML algorithms require numerical input, encoding is essential for handling categorical variables effectively.  

**Usefulness in Data Science:**  
- Helps machine learning models understand and process categorical data.  
- Improves model accuracy by making features interpretable by mathematical operations.  
- Enables better feature selection and enhances model performance.  

---

### **Q2. What is nominal encoding? Provide an example of how you would use it in a real-world scenario.**  
**Answer:**  
Nominal encoding is used to represent categorical data where there is no inherent order among the categories. It assigns a unique integer to each category.  

**Example:**  
Suppose you are analyzing different car brands in a dataset:  
**Car Brand:** Toyota, Honda, Ford, BMW, Tesla  

Using nominal encoding, they can be represented as:  
- Toyota → 0  
- Honda → 1  
- Ford → 2  
- BMW → 3  
- Tesla → 4  

This transformation allows machine learning models to process the data numerically.

---

### **Q3. In what situations is nominal encoding preferred over one-hot encoding? Provide a practical example.**  
**Answer:**  
Nominal encoding is preferred over one-hot encoding when:  
- The categorical variable has many unique values (to avoid high-dimensionality issues).  
- There is no meaningful ordinal relationship between categories.  

**Example:**  
Consider a dataset with 500 unique cities. One-hot encoding would create 500 columns, which is inefficient and leads to sparse data. Instead, using nominal encoding (assigning a unique integer to each city) reduces the feature space significantly while maintaining model performance.

### **Q4. Suppose you have a dataset containing categorical data with 5 unique values. Which encoding technique would you use to transform this data into a format suitable for machine learning algorithms? Explain why you made this choice.**  
**Answer:**  
If the categorical data has 5 unique values, the choice of encoding depends on the dataset size and algorithm:  
- **One-Hot Encoding:** If the dataset is small and the number of categories is manageable (e.g., ≤10).  
- **Ordinal/Nominal Encoding:** If there is no inherent order and the dataset is large to prevent the curse of dimensionality.  

For a dataset with 5 unique values, **one-hot encoding** would be preferable for models like Decision Trees and Neural Networks, while **nominal encoding** would be better for linear models like Logistic Regression.

### **Q5. In a machine learning project, you have a dataset with 1000 rows and 5 columns. Two of the columns are categorical, and the remaining three columns are numerical. If you were to use nominal encoding to transform the categorical data, how many new columns would be created? Show your calculations.**  
**Answer:**  
If both categorical columns have **n1** and **n2** unique values, nominal encoding replaces them with integer labels (one column per categorical feature).  

- Let’s assume categorical column 1 has **4 unique values**.  
- Categorical column 2 has **5 unique values**.  
- Nominal encoding assigns one integer per category, so the number of new columns remains **2** (one for each categorical column).  

Thus, the dataset will have **(3 + 2) = 5 columns** in total.

### **Q6. You are working with a dataset containing information about different types of animals, including their species, habitat, and diet. Which encoding technique would you use to transform the categorical data into a format suitable for machine learning algorithms? Justify your answer.**  
**Answer:**  
For a dataset with categorical features like species, habitat, and diet:  
- **One-Hot Encoding** is preferred if the number of categories is small (e.g., 5–10 unique values per column).  
- **Nominal Encoding** is used if there are many unique values (e.g., 100+ species) to avoid increasing the dimensionality too much.  

Since species, habitat, and diet do not have an ordinal relationship, one-hot encoding is a better choice for tree-based models, whereas nominal encoding is efficient for linear models.

### **Q7. You are working on a project that involves predicting customer churn for a telecommunications company. You have a dataset with 5 features, including the customer's gender, age, contract type, monthly charges, and tenure. Which encoding technique(s) would you use to transform the categorical data into numerical data? Provide a step-by-step explanation of how you would implement the encoding.**  
**Answer:**  
The dataset consists of:  
- **Categorical Features:** Gender, Contract Type  
- **Numerical Features:** Age, Monthly Charges, Tenure  

**Step-by-Step Encoding Process:**  
1. **Identify categorical columns:**  
   - `Gender`: Binary categorical (Male/Female) → **Binary Encoding (0 or 1)**  
   - `Contract Type`: Multiple categories (e.g., Monthly, Yearly, Two-Year) → **One-Hot Encoding**  

2. **Apply Encoding Techniques:**  
   - Convert `Gender` using binary encoding:  
     - Male → 0, Female → 1  
   - Apply **one-hot encoding** for `Contract Type`:  
     - Monthly → [1, 0, 0]  
     - Yearly → [0, 1, 0]  
     - Two-Year → [0, 0, 1]  

3. **Concatenate Encoded Data:**  
   - Merge transformed categorical data with numerical features.  

4. **Ensure Model Compatibility:**  
   - Normalize numerical features if necessary (e.g., Min-Max Scaling).  

This ensures the dataset is properly formatted for machine learning algorithms.
