
### **Q1. Load the "titanic" dataset using the `load_dataset` function of Seaborn. Use Plotly Express to plot a scatter plot for the age and fare columns in the Titanic dataset.**


---

```python
import seaborn as sns
import plotly.express as px

# Load the Titanic dataset
titanic = sns.load_dataset("titanic")

# Create a scatter plot using Plotly Express
fig = px.scatter(titanic, x="age", y="fare", title="Scatter Plot of Age vs Fare in Titanic Dataset",
                 labels={"age": "Age", "fare": "Fare"})
fig.show()
```

### **Q2. Using the tips dataset in the Plotly library, plot a box plot using Plotly Express.**

---

```python
import plotly.express as px

# Load the tips dataset from Plotly
tips = px.data.tips()

# Create a box plot
fig = px.box(tips, x="day", y="total_bill", color="sex", title="Box Plot of Total Bill by Day & Gender")
fig.show()
```


### **Q3. Using the tips dataset in the Plotly library, plot a histogram for `x="sex"` and `y="total_bill"` column. Also, use the `smoker` column with the `pattern_shape` parameter and the `day` column with the `color` parameter.**


---

```python
import plotly.express as px

# Load the tips dataset
tips = px.data.tips()

# Create a histogram
fig = px.histogram(tips, x="sex", y="total_bill", color="day", pattern_shape="smoker",
                   title="Histogram of Total Bill by Gender with Smoker & Day Variation")
fig.show()
```


### **Q4. Using the iris dataset in the Plotly library, plot a scatter matrix plot using the `species` column for the `color` parameter.**


```python
import plotly.express as px

# Load the iris dataset
iris = px.data.iris()

# Create a scatter matrix
fig = px.scatter_matrix(iris, dimensions=["sepal_length", "sepal_width", "petal_length", "petal_width"],
                        color="species", title="Scatter Matrix of Iris Dataset")
fig.show()
```


### **Q5. What is Distplot? Using Plotly Express, plot a distplot.**
**Answer:**  
A **Distplot (Distribution Plot)** is a visualization that shows the distribution of a dataset. It combines a **histogram** and a **KDE (Kernel Density Estimation) plot**, helping to understand the distribution of numerical data.

**Plotly Express does not have a built-in `distplot`**, but we can use `plotly.figure_factory` to achieve a similar result.


```python
import plotly.figure_factory as ff
import numpy as np

# Generate random data
data = [np.random.randn(200)]

# Create a distplot
fig = ff.create_distplot(data, group_labels=["Random Data"], bin_size=0.2, show_rug=True)
fig.update_layout(title="Distribution Plot (Distplot) using Plotly")
fig.show()
```
