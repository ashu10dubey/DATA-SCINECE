### Q1. What is the Filter method in feature selection, and how does it work?  
**Answer:**  
The **Filter method** is a feature selection technique that evaluates the relevance of each feature based on statistical measures before training a machine learning model. It does not depend on any specific model but instead relies on intrinsic properties of the data, such as correlation, variance, or mutual information.  
**How it works:**  
1. Compute statistical scores for each feature using techniques like **Chi-square test, ANOVA, correlation coefficients, or mutual information**.  
2. Rank the features based on their scores.  
3. Select the top-ranked features according to a predefined threshold.  

---

### Q2. How does the Wrapper method differ from the Filter method in feature selection?  
**Answer:**  
The **Wrapper method** differs from the **Filter method** in the following ways:  
- **Wrapper Method:** Uses a machine learning model to evaluate different feature subsets by training and testing the model on various combinations of features. It is computationally expensive but usually provides better feature selection results.  
- **Filter Method:** Selects features based on statistical properties without involving a model. It is computationally efficient but may overlook important interactions between features.  

---

### Q3. What are some common techniques used in Embedded feature selection methods?  
**Answer:**  
**Embedded methods** incorporate feature selection as part of the model training process. Common techniques include:  
1. **Lasso (L1 Regularization)** – Shrinks less important features' coefficients to zero.  
2. **Decision Trees & Random Forests** – Uses feature importance scores to rank and select features.  
3. **Gradient Boosting Methods (e.g., XGBoost, LightGBM)** – Assigns importance scores to features.  
4. **Recursive Feature Elimination (RFE) with models like SVM or Logistic Regression** – Iteratively removes least important features.  

---

### Q4. What are some drawbacks of using the Filter method for feature selection?  
**Answer:**  
1. **Ignores Feature Interactions:** It evaluates each feature individually without considering relationships between them.  
2. **May Remove Useful Features:** Some features might appear unimportant statistically but can contribute significantly when combined with others.  
3. **Not Optimized for a Specific Model:** It does not consider how selected features impact model performance.  

---

### Q5. In which situations would you prefer using the Filter method over the Wrapper method for feature selection?  
**Answer:**  
- When **working with large datasets** where computational efficiency is important.  
- When **domain knowledge** suggests some features should be retained based on statistical properties.  
- When performing **preliminary feature selection** before applying more complex techniques.  
- When **model training time is a constraint** and wrapper methods would be too slow.

### Q6. In a telecom company, you are working on a project to develop a predictive model for customer churn. You are unsure of which features to include in the model because the dataset contains several different ones. Describe how you would choose the most pertinent attributes for the model using the Filter Method.  
**Answer:**  
Using the **Filter Method**, I would:  
1. **Remove Irrelevant or Low-Variance Features** – Drop features that have little variability (e.g., constant values).  
2. **Apply Correlation Analysis** – Remove highly correlated features (retain only one from each correlated pair).  
3. **Use Chi-Square Test or Mutual Information** – Select categorical features with the highest association with the target (churn).  
4. **Use ANOVA (Analysis of Variance)** – Identify the numerical features that have significant impact on the target.  
5. **Rank Features & Select the Best Ones** – Keep the top-ranked features based on the above tests.  
6. **Validate with a Model** – Use a simple model (e.g., Logistic Regression) to check if the filtered features improve predictive performance.

### Q7. You are working on a project to predict the outcome of a soccer match. You have a large dataset with many features, including player statistics and team rankings. Explain how you would use the Embedded method to select the most relevant features for the model.  
**Answer:**  
To use the **Embedded Method**, I would:  
1. **Train a Decision Tree or Random Forest Model** – These models calculate feature importance based on how much they reduce prediction errors.  
2. **Use Lasso Regression (L1 Regularization)** – Assign zero coefficients to irrelevant features and retain only significant ones.  
3. **Apply Gradient Boosting (XGBoost, LightGBM)** – These models automatically rank and select important features.  
4. **Perform Recursive Feature Elimination (RFE)** – Start with all features, iteratively remove the least important ones, and keep refining the selection.  
5. **Validate with Model Performance** – Use cross-validation to ensure that the selected features improve prediction accuracy.

### Q8. You are working on a project to predict the price of a house based on its features, such as size, location, and age. You have a limited number of features, and you want to ensure that you select the most important ones for the model. Explain how you would use the Wrapper method to select the best set of features for the predictor.  
**Answer:**  
Using the **Wrapper Method**, I would:  
1. **Define a Base Model** – Use a regression model like **Linear Regression, Decision Tree, or Random Forest**.  
2. **Use Forward Selection:**  
   - Start with no features and add them one by one, keeping the feature that improves the model the most.  
3. **Use Backward Elimination:**  
   - Start with all features and remove the least important one in each iteration.  
4. **Perform Recursive Feature Elimination (RFE):**  
   - Train the model with all features, rank them by importance, and iteratively remove the least significant features.  
5. **Evaluate Model Performance** – Use cross-validation to ensure that the selected features improve the house price prediction model.
