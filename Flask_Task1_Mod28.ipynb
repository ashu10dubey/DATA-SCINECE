### **Q1. What is Flask Framework? What are the advantages of Flask Framework?**  

#### **Flask Framework:**
Flask is a **lightweight web framework** for Python that is used to build web applications. It is **based on WSGI (Web Server Gateway Interface)** and **Jinja2 templating engine**. It is widely used for **building APIs and web services**.

#### **Advantages of Flask Framework:**
1. **Lightweight and Simple:** Minimal dependencies and easy to use.  
2. **Modular and Scalable:** Can be extended with additional libraries.  
3. **Built-in Development Server & Debugger:** Helps in testing.  
4. **Jinja2 Templating Support:** Allows dynamic HTML rendering.  
5. **URL Routing:** Enables mapping of URLs to functions.  
6. **RESTful Request Handling:** Easy to create APIs.  
7. **Database Integration:** Supports SQLAlchemy for ORM.  
8. **Easy Deployment:** Can be deployed on cloud platforms like AWS, Heroku.

### **Q2. Create a simple Flask application to display ‘Hello World!!’. Attach the screenshot of the output in Jupyter Notebook.**  

#### **Python Code (Flask App)**
```python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def hello():
    return "Hello World!!"

if __name__ == "__main__":
    app.run(debug=True)
```
### **Steps to Run the Flask App**
1. Save this Python script as `app.py`.
2. Run the script using the command:  
   ```
   python app.py
   ```
3. Open a web browser and visit:  
   ```
   http://127.0.0.1:5000/
   ```

### **Q3. What is App Routing in Flask? Why do we use app routes?**  

#### **App Routing in Flask**
App Routing refers to **binding a URL to a specific function** so that when a user visits the URL, the corresponding function executes.

#### **Why use app routes?**
- To **define different pages** in the web application.  
- To make the application **modular and structured**.  
- To handle **multiple URLs and HTTP methods**.  
- To create **RESTful APIs**.  

#### **Example of Routing in Flask**
```python
@app.route("/")  # Route for Home Page
def home():
    return "Welcome to Home Page"

@app.route("/about")  # Route for About Page
def about():
    return "About Us Page"
```

### **Q4. Create a “/welcome” route to display the welcome message and a “/” route to show company details.**  

#### **Python Code (Flask App)**
```python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return """
    <h1>Company Name: ABC Corporation</h1>
    <h2>Location: India</h2>
    <h3>Contact Detail: 999-999-9999</h3>
    """

@app.route("/welcome")
def welcome():
    return "<h1>Welcome to ABC Corporation</h1>"

if __name__ == "__main__":
    app.run(debug=True)
```

### **Steps to Run:**
1. Save the script as `app.py`.
2. Run `python app.py`.
3. Visit:
   - `http://127.0.0.1:5000/` → Displays Company Details.
   - `http://127.0.0.1:5000/welcome` → Displays "Welcome to ABC Corporation".

### **Q5. What function is used in Flask for URL Building? Write a Python code to demonstrate the working of the url_for() function.**  

#### **Function Used for URL Building:**
- The **`url_for()` function** is used to build URLs dynamically instead of hardcoding them.

#### **Example of `url_for()` Function:**
```python
from flask import Flask, url_for

app = Flask(__name__)

@app.route("/")
def home():
    return "This is the Home Page."

@app.route("/about")
def about():
    return "This is the About Page."

@app.route("/contact")
def contact():
    return "Contact Us at: 999-999-9999"

@app.route("/links")
def links():
    return f"""
    <a href="{url_for('home')}">Home</a><br>
    <a href="{url_for('about')}">About</a><br>
    <a href="{url_for('contact')}">Contact</a><br>
    """

if __name__ == "__main__":
    app.run(debug=True)
```

### **Steps to Run:**
1. Save the script as `app.py`.
2. Run `python app.py`.
3. Visit `http://127.0.0.1:5000/links` → You will see **dynamic links** to different pages.
