
### **Q1. What is MongoDB? Explain non-relational databases in short. In which scenarios is it preferred to use MongoDB over SQL databases?**  

**MongoDB** is a **NoSQL (non-relational) database** that stores data in **JSON-like BSON (Binary JSON) format**. It is **document-oriented**, meaning data is stored in flexible documents rather than tables.  

#### **Non-relational databases (NoSQL) in short:**
- Do **not use fixed schema** like SQL databases.
- Handle **unstructured or semi-structured** data efficiently.
- Scale **horizontally** (across multiple machines).
- Provide **high availability and replication**.

#### **When to use MongoDB over SQL databases?**
- When handling **large volumes of unstructured data**.
- When data **structure changes frequently**.
- When fast **read/write operations** are required.
- When handling **big data applications** (e.g., IoT, analytics, content management).
- When building **real-time applications** (e.g., chat applications, recommendation systems).

### **Q2. State and Explain the features of MongoDB.**  

MongoDB offers several **powerful features**:  

1. **Document-Oriented Storage** → Uses **BSON (Binary JSON)** instead of tables.  
2. **Schema-less** → No predefined schema, allowing **flexibility** in data storage.  
3. **Scalability** → Supports **horizontal scaling (Sharding)** for handling large datasets.  
4. **High Performance** → Provides **indexing, replication, and caching** for faster operations.  
5. **Replication (High Availability)** → Uses **Replica Sets** to ensure data redundancy.  
6. **Aggregation Framework** → Provides **powerful querying and data transformation**.  
7. **Automatic Failover** → Automatically recovers from **server failures**.  
8. **Rich Query Language** → Supports **filtering, sorting, and indexing** like SQL.

### **Q4. Insert one record, insert multiple records, and retrieve data using find() and find_one().**  

```python
import pymongo  

# Connect to MongoDB
client = pymongo.MongoClient("mongodb://localhost:27017/")  
db = client["my_database"]  
collection = db["my_collection"]  

# Insert One Record
record = {"name": "Ashutosh", "age": 23, "city": "Bangalore"}  
collection.insert_one(record)  
print("One Record Inserted!")  

# Insert Multiple Records
records = [
    {"name": "Amit", "age": 25, "city": "Delhi"},
    {"name": "Ravi", "age": 27, "city": "Mumbai"},
    {"name": "Sita", "age": 22, "city": "Chennai"}
]
collection.insert_many(records)  
print("Multiple Records Inserted!")  

# Retrieve One Record
print("Find One:", collection.find_one())  

# Retrieve All Records
print("Find All:")
for record in collection.find():
    print(record)
```

---

### **Q5. How does the find() method work in MongoDB? Write an example.**  

The `find()` method **retrieves multiple records** from a MongoDB collection.  

```python
# Retrieve all records in the collection
for record in collection.find():
    print(record)

# Retrieve specific fields (only name and age)
for record in collection.find({}, {"_id": 0, "name": 1, "age": 1}):
    print(record)

# Retrieve records with age greater than 23
for record in collection.find({"age": {"$gt": 23}}):
    print(record)
```

### **Q6. Explain the sort() method. Demonstrate sorting in MongoDB.**  

The `sort()` method **arranges data** in ascending or descending order.  
- **Ascending Order:** `sort({field: 1})`  
- **Descending Order:** `sort({field: -1})`  

#### **Example:**
```python
# Sort records in Ascending Order (by age)
for record in collection.find().sort("age", 1):
    print(record)

# Sort records in Descending Order (by age)
for record in collection.find().sort("age", -1):
    print(record)
```

### **Q7. Explain delete_one(), delete_many(), and drop() methods.**  

#### **1️⃣ `delete_one()` → Deletes a single matching document.**  
```python
collection.delete_one({"name": "Ashutosh"})
```
- Deletes **only the first matching record**.

#### **2️⃣ `delete_many()` → Deletes multiple matching documents.**  
```python
collection.delete_many({"age": {"$gt": 25}})
```
- Deletes **all documents where age > 25**.

#### **3️⃣ `drop()` → Deletes an entire collection.**  
```python
collection.drop()
```
- **Removes the collection permanently** from the database.
