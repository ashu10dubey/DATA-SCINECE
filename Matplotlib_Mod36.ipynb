

### **Q1: What is Matplotlib? Why is it used? Name five plots that can be plotted using the Pyplot module of Matplotlib.**
**Answer:**
Matplotlib is a widely used plotting library in Python for creating static, animated, and interactive visualizations. It provides various tools to create graphs and charts that help in data analysis and visualization.

**Why is it used?**
- To visualize data and gain insights.
- To generate publication-quality plots.
- To customize plots easily (colors, labels, legends, etc.).
- To create a variety of charts like line plots, bar plots, scatter plots, etc.

**Five plots that can be created using `pyplot` module:**
1. Line Plot
2. Scatter Plot
3. Bar Plot
4. Histogram
5. Box Plot

### **Q2: What is a scatter plot?**
A scatter plot is used to display values for two variables. It helps in identifying relationships, trends, or correlations between variables.

**Code to generate a scatter plot:**
```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(3)
x = 3 + np.random.normal(0, 2, 50)
y = 3 + np.random.normal(0, 2, len(x))

# Scatter Plot
plt.scatter(x, y, color='blue', marker='o', label='Data points')
plt.title('Scatter Plot of Generated Data')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.grid()
plt.show()
```

### **Q3: Why is the subplot() function used?**
The `subplot()` function in Matplotlib is used to create multiple plots in the same figure. It allows us to organize multiple visualizations efficiently.

**Code to draw four line plots using subplot():**
```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3, 4, 5])

y1 = np.array([0, 100, 200, 300, 400, 500])
y2 = np.array([50, 20, 40, 20, 60, 70])
y3 = np.array([10, 20, 30, 40, 50, 60])
y4 = np.array([200, 350, 250, 550, 450, 150])

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.plot(x, y1, color='blue', marker='o')
plt.title("Line Plot 1")

plt.subplot(2, 2, 2)
plt.plot(x, y2, color='red', linestyle='--')
plt.title("Line Plot 2")

plt.subplot(2, 2, 3)
plt.plot(x, y3, color='green', marker='s')
plt.title("Line Plot 3")

plt.subplot(2, 2, 4)
plt.plot(x, y4, color='purple', linestyle=':')
plt.title("Line Plot 4")

plt.tight_layout()
plt.show()
```

### **Q4: What is a bar plot? Why is it used?**
A bar plot is a graphical representation of categorical data using rectangular bars. It is used to compare different categories.

**Code to create bar plots:**
```python
import numpy as np
import matplotlib.pyplot as plt

company = np.array(["Apple", "Microsoft", "Google", "AMD"])
profit = np.array([3000, 8000, 1000, 10000])

# Vertical Bar Plot
plt.bar(company, profit, color=['blue', 'green', 'red', 'purple'])
plt.title('Company Profits')
plt.xlabel('Company')
plt.ylabel('Profit in Millions')
plt.show()

# Horizontal Bar Plot
plt.barh(company, profit, color=['blue', 'green', 'red', 'purple'])
plt.title('Company Profits (Horizontal)')
plt.xlabel('Profit in Millions')
plt.ylabel('Company')
plt.show()
```

### **Q5: What is a box plot? Why is it used?**
A box plot (or whisker plot) is used to visualize the distribution of numerical data and highlight outliers. It provides insights into data spread, median, quartiles, and possible outliers.

**Code to create a box plot:**
```python
import numpy as np
import matplotlib.pyplot as plt

box1 = np.random.normal(100, 10, 200)
box2 = np.random.normal(90, 20, 200)

data = [box1, box2]

plt.boxplot(data, labels=['Dataset 1', 'Dataset 2'], patch_artist=True)
plt.title('Box Plot Example')
plt.xlabel('Dataset')
plt.ylabel('Values')
plt.grid()
plt.show()
```
