### **Q1. What is the purpose of forward propagation in a neural network?**  
Forward propagation computes the **predicted output** by passing inputs through the network to calculate the **loss function** for optimization.

---

### **Q2. How is forward propagation implemented mathematically in a single-layer feedforward neural network?**  
For a **single-layer** feedforward network:

$$ Z = W X + b $$

$$ A = f(Z) $$

where:  
- \( X \) = Input vector  
- \( W \) = Weight matrix  
- \( b \) = Bias vector  
- \( Z \) = Weighted sum  
- \( A \) = Output after activation function \( f(Z) \)

For **multi-layer networks**, this is repeated for each layer.

---

### **Q3. How are activation functions used during forward propagation?**  
Activation functions apply **non-linearity** to help the network learn complex patterns:

$$ A = f(Z) $$

Examples:  
- **Sigmoid:** $$ \sigma(x) = \frac{1}{1 + e^{-x}} $$  
- **ReLU:** $$ f(x) = \max(0, x) $$  
- **Tanh:** $$ f(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} $$  

---

### **Q4. What is the role of weights and biases in forward propagation?**  
- **Weights (\( W \))** control how much influence each input has.  
- **Biases (\( b \))** shift activation thresholds.  
- The equation:  

$$ Z = W X + b $$  

ensures the network learns meaningful patterns.

---

### **Q5. What is the purpose of applying a softmax function in the output layer during forward propagation?**  
Softmax converts raw scores into **probabilities** for multi-class classification:

$$ \sigma(z_i) = \frac{e^{z_i}}{\sum_{j} e^{z_j}} $$

where:  
- \( z_i \) = Logit score for class \( i \)  
- \( \sigma(z_i) \) = Probability of class \( i \)  
- Ensures all probabilities sum to **1**.

---

### **Q6. What is the purpose of backward propagation in a neural network?**  
Backpropagation minimizes error by updating weights and biases using **gradient descent**.

---

### **Q7. How is backward propagation mathematically calculated in a single-layer feedforward neural network?**  
Loss function \( L \):

$$ L = \frac{1}{m} \sum_{i=1}^{m} \text{Loss}(\hat{Y}_i, Y_i) $$

Gradients:  

1. Compute gradient w.r.t. activation output:  

   $$ \frac{\partial L}{\partial A} $$  

2. Compute gradient w.r.t. weighted sum:  

   $$ \frac{\partial L}{\partial Z} = \frac{\partial L}{\partial A} \cdot f'(Z) $$  

3. Compute gradient w.r.t. weights and biases:  

   $$ \frac{\partial L}{\partial W} = \frac{\partial L}{\partial Z} \cdot X^T $$  

   $$ \frac{\partial L}{\partial b} = \sum \frac{\partial L}{\partial Z} $$  

4. Update parameters:  

   $$ W = W - \alpha \frac{\partial L}{\partial W} $$  

   $$ b = b - \alpha \frac{\partial L}{\partial b} $$  

where \( \alpha \) is the **learning rate**.

---

### **Q8. Can you explain the concept of the chain rule and its application in backward propagation?**  
The **chain rule** is used to compute gradients in deep networks:

$$ \frac{dL}{dX} = \frac{dL}{dZ} \cdot \frac{dZ}{dX} $$

For backpropagation:

$$ \frac{\partial L}{\partial W} = \frac{\partial L}{\partial A} \cdot \frac{\partial A}{\partial Z} \cdot \frac{\partial Z}{\partial W} $$

This allows adjusting **weights and biases** efficiently.

---

### **Q9. What are some common challenges or issues that can occur during backward propagation, and how can they be addressed?**  
1. **Vanishing Gradient** – Small gradients slow learning.  
   - **Solution:** Use **ReLU** or **Batch Normalization**.  
2. **Exploding Gradient** – Large gradients cause instability.  
   - **Solution:** Apply **gradient clipping**.  
3. **Overfitting** – The model memorizes training data.  
   - **Solution:** Apply **dropout** or **L2 regularization**.  
4. **Slow Convergence** – Poor weight initialization.  
   - **Solution:** Use **Xavier/He initialization**.
