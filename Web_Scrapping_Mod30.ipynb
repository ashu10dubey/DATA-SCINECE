### **Q1. What is Web Scraping? Why is it Used? Give three areas where Web Scraping is used to get data.**  
#### **What is Web Scraping?**  
Web Scraping is the process of extracting data from websites using automated tools or scripts. The extracted data is then stored for analysis, automation, or use in other applications.  

#### **Why is Web Scraping Used?**  
- To collect large amounts of data quickly.  
- To automate the process of extracting information from dynamic websites.  
- To gather structured data from unstructured web pages.  

#### **Three Areas Where Web Scraping is Used:**  
1. **E-commerce & Price Monitoring** – Extracting product prices, reviews, and inventory status from competitor websites.  
2. **News & Sentiment Analysis** – Collecting news articles, blog posts, and social media data for trend analysis.  
3. **Real Estate & Job Listings** – Gathering real estate property details or job openings from various websites.

### **Q2. What are the different methods used for Web Scraping?**  
The primary methods for Web Scraping include:  

1. **Using BeautifulSoup (HTML Parsing):**  
   - Extracts structured data from HTML and XML files.  
   - Works well for simple static websites.  

2. **Using Selenium (Browser Automation):**  
   - Automates browsing and interacts with JavaScript-based websites.  
   - Useful for scraping dynamic content.  

3. **Using Scrapy (Web Crawling Framework):**  
   - Suitable for large-scale web scraping and crawling tasks.  
   - Supports asynchronous requests for efficiency.  

4. **Using APIs (If Available):**  
   - Many websites provide public APIs to access data legally.  
   - Faster and more structured than direct web scraping.

### **Q3. What is BeautifulSoup? Why is it used?**  
- **BeautifulSoup** is a Python library used for extracting data from HTML and XML files.  
- It allows users to navigate the **DOM tree**, search for elements, and modify them easily.  

#### **Why is BeautifulSoup Used?**  
- It **parses and processes HTML data efficiently**.  
- It provides methods like `.find()`, `.find_all()`, and `.select()` to extract specific elements.  
- Works well with static websites and requires minimal setup.  

#### **Example Code:**  
```python
from bs4 import BeautifulSoup
import requests

url = "https://example.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Extracting all links
links = soup.find_all('a')
for link in links:
    print(link['href'])
```

### **Q4. Why is Flask used in this Web Scraping project?**  
- **Flask** is used to create a **web interface** for displaying scraped data.  
- It allows integration with REST APIs and **MongoDB** for data storage.  
- Flask makes it easy to deploy the project on a web server or cloud platform.  

#### **How Flask Helps in Web Scraping:**  
1. **API Endpoint Creation:** Flask serves scraped data through REST APIs.  
2. **Rendering Data on Web Pages:** Scraped data can be displayed dynamically on web portals.  
3. **Integration with Databases:** Flask allows storing scraped data in **MongoDB**.  
4. **Deployment on AWS or Cloud:** Flask apps can be easily hosted on **AWS EC2** or **AWS Lambda**.

### **Q5. Write the names of AWS services used in this project. Also, explain the use of each service.**  

#### **AWS Services Used:**
1. **Amazon EC2 (Elastic Compute Cloud):**  
   - Provides virtual servers to host Flask applications.  
   - Used for running web scraping scripts on a dedicated server.  

2. **Amazon S3 (Simple Storage Service):**  
   - Stores large datasets collected from web scraping.  
   - Useful for storing scraped images, CSVs, and JSON files.  

3. **Amazon RDS (Relational Database Service) / Amazon DynamoDB:**  
   - Stores structured data (if using SQL databases).  
   - DynamoDB is a NoSQL alternative for scalable data storage.  

4. **AWS Lambda (Serverless Functions):**  
   - Automates web scraping tasks without managing a server.  
   - Can run scraping scripts periodically (e.g., every 24 hours).  

5. **Amazon CloudWatch:**  
   - Monitors logs and performance metrics of scraping scripts.  
   - Alerts on failures or high resource usage.  

#### **Example AWS Deployment Workflow:**  
1. Scrape data using **BeautifulSoup or Selenium**.  
2. Store the scraped data in **MongoDB (hosted on EC2 or DynamoDB)**.  
3. Use **Flask** to expose the data via a REST API.  
4. Deploy the Flask app on **AWS EC2**.  
5. Schedule periodic scraping using **AWS Lambda**.
