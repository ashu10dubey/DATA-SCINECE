### **Q1. What is the purpose of grid search CV in machine learning, and how does it work?**  
Grid Search Cross-Validation (**GridSearchCV**) is used to **find the best hyperparameters** by evaluating all possible combinations from a predefined grid.  

**How it works:**  
1. Define a **grid** of hyperparameter values.  
2. Train and validate the model on each combination using **k-fold cross-validation**.  
3. Select the best hyperparameters based on performance metrics (e.g., accuracy, F1-score).  

---

### **Q2. Describe the difference between grid search CV and randomized search CV, and when might you choose one over the other?**  
- **GridSearchCV:**  
  - Exhaustively evaluates **all** possible hyperparameter combinations.  
  - **Best for small search spaces** but computationally expensive.  

- **RandomizedSearchCV:**  
  - Randomly samples a subset of hyperparameters from the grid.  
  - **Best for large search spaces** with limited computational resources.  

**When to use which?**  
- **GridSearchCV** → When precision matters, and computational resources are available.  
- **RandomizedSearchCV** → When the search space is large, and we need a quick approximation.

### **Q3. What is data leakage, and why is it a problem in machine learning? Provide an example.**  
**Data leakage** occurs when information from outside the training dataset **unintentionally** influences the model, leading to **overoptimistic** performance.  

**Example:**  
- Including **future information** in training data, such as using test labels during preprocessing.  

---

### **Q4. How can you prevent data leakage when building a machine learning model?**  
1. **Split data properly:** Ensure training, validation, and test sets are truly independent.  
2. **Feature engineering after splitting:** Avoid using test data information while engineering features.  
3. **Pipeline approach:** Use `sklearn.pipeline` to apply transformations within cross-validation.  
4. **Be cautious with time series data:** Use past data only for training to prevent **future leakage**.

### **Q5. What is a confusion matrix, and what does it tell you about the performance of a classification model?**  
A **confusion matrix** is a table used to **evaluate classification performance** by comparing actual vs. predicted labels.  

\[
\begin{bmatrix}
TP & FP \\
FN & TN
\end{bmatrix}
\]

where:  
- **TP** = True Positives  
- **FP** = False Positives  
- **FN** = False Negatives  
- **TN** = True Negatives  

It helps analyze **errors**, **biases**, and overall **classification performance**.  

---

### **Q6. Explain the difference between precision and recall in the context of a confusion matrix.**  
- **Precision** (Positive Predictive Value):  
  - Measures how many predicted positives are actually correct.  
  - Formula:  

    $$ \text{Precision} = \frac{TP}{TP + FP} $$  

- **Recall** (Sensitivity):  
  - Measures how many actual positives were correctly identified.  
  - Formula:  

    $$ \text{Recall} = \frac{TP}{TP + FN} $$  

**Trade-off:**  
- **High Precision, Low Recall** → Model is conservative (few false positives).  
- **High Recall, Low Precision** → Model is aggressive (few false negatives).  

---

### **Q7. How can you interpret a confusion matrix to determine which types of errors your model is making?**  
- **High FP (False Positives):** Model is over-predicting positives.  
- **High FN (False Negatives):** Model is missing true positives.  
- **Balanced FN and FP:** Model is well-calibrated.  

**Example:**  
- In fraud detection, **FN (missed fraud cases)** is more critical than **FP (wrongly flagged cases)**.  

---

### **Q8. What are some common metrics that can be derived from a confusion matrix, and how are they calculated?**  

1. **Accuracy**:  
   $$ \text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN} $$  

2. **Precision**:  
   $$ \text{Precision} = \frac{TP}{TP + FP} $$  

3. **Recall**:  
   $$ \text{Recall} = \frac{TP}{TP + FN} $$  

4. **F1-Score** (Harmonic mean of Precision & Recall):  
   $$ \text{F1} = \frac{2 \times \text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} $$  

5. **Specificity (True Negative Rate):**  
   $$ \text{Specificity} = \frac{TN}{TN + FP} $$  

---

### **Q9. What is the relationship between the accuracy of a model and the values in its confusion matrix?**  
**Accuracy** is calculated from the confusion matrix as:  

$$ \text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN} $$  

However, accuracy can be **misleading** in imbalanced datasets.  
- **Example:** If 95% of data is "Negative" and model always predicts "Negative", accuracy = 95% but **recall for positive class is 0%**.  

---

### **Q10. How can you use a confusion matrix to identify potential biases or limitations in your machine learning model?**  
- **Imbalance in FP/FN values:** Indicates class imbalance.  
- **Higher FN than FP:** Model is cautious but might be missing important cases (e.g., **cancer diagnosis**).  
- **Higher FP than FN:** Model is aggressive in predicting positives (e.g., **spam detection**).  
- **Similar FN and FP:** Well-balanced model.
