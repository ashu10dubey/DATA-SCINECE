**Q1. What are the Probability Mass Function (PMF) and Probability Density Function (PDF)? Explain with an example.**

**A1.**

* **Probability Mass Function (PMF):**
    - The PMF is used for **discrete** random variables.
    - It gives the probability that a discrete random variable takes on a specific value.
    - Example:
        - Consider the roll of a fair six-sided die. The PMF of the random variable "X" representing the outcome of the roll is:
            - P(X = 1) = 1/6
            - P(X = 2) = 1/6
            - P(X = 3) = 1/6
            - P(X = 4) = 1/6
            - P(X = 5) = 1/6
            - P(X = 6) = 1/6

* **Probability Density Function (PDF):**
    - The PDF is used for **continuous** random variables.
    - It describes the relative likelihood of a continuous random variable taking on a given value.
    - The probability of the variable falling within a particular interval is given by the area under the PDF curve over that interval.
    - Example:
        - The PDF of a standard normal distribution is given by:
            - f(x) = (1 / sqrt(2 * pi)) * exp(-x^2 / 2)
        - The PDF itself does not represent the probability of the variable taking on a specific value directly, but rather the probability *density* at that point.

**Q2. What is Cumulative Density Function (CDF)? Explain with an example. Why CDF is used?**

**A2.**

* **Cumulative Density Function (CDF):**
    - The CDF of a random variable (discrete or continuous) gives the probability that the variable will take a value less than or equal to a specific value.
    - It is a non-decreasing function that ranges from 0 to 1.

    - **Example:**
        - For a standard normal distribution, the CDF gives the probability that a randomly drawn value from that distribution will be less than or equal to a specific value 'x'.

* **Why CDF is used:**
    - **Calculating probabilities:** CDF allows you to easily calculate the probability of the random variable falling within a certain range. For example, to find the probability that a variable X lies between a and b, you can calculate CDF(b) - CDF(a).
    - **Quantile calculations:** CDF can be used to find the value of the variable corresponding to a given probability (e.g., finding the 95th percentile).
    - **Comparing distributions:** CDFs can be used to compare different probability distributions.

**Q6. Consider a dataset with a mean of 50 and a standard deviation of 10. If we assume that the dataset
is normally distributed, what is the probability that a randomly selected observation will be greater
than 60? Use the appropriate formula and show your calculations.**

**A6.**

1. **Standardize the value:**
   - Calculate the z-score: z = (x - μ) / σ = (60 - 50) / 10 = 1

2. **Find the probability using a z-table:**
   - Look up the z-score (1) in a standard normal distribution table. This will give you the probability of getting a value less than 60.
   - The probability of getting a value less than 60 is approximately 0.8413.

3. **Calculate the probability of getting a value greater than 60:**
   - P(X > 60) = 1 - P(X <= 60) = 1 - 0.8413 = 0.1587

Therefore, the probability that a randomly selected observation will be greater than 60 is approximately **0.1587**.

**Q8. What is the z score? State the importance of the z score.**

**A8.**

* **Z-score:**
    - A measure of how many standard deviations a data point is from the mean of a distribution.
    - Calculated as: z = (x - μ) / σ
    - Where:
        - x is the data point
        - μ is the mean of the distribution
        - σ is the standard deviation of the distribution

* **Importance of z-score:**
    - **Standardization:** Z-scores transform data to a standard normal distribution with a mean of 0 and a standard deviation of 1. This allows for easy comparison of data points from different distributions.
    - **Outlier detection:** Z-scores can help identify outliers by identifying data points that are significantly far from the mean.
    - **Statistical inference:** Z-scores are used in many statistical tests and calculations, such as hypothesis testing and confidence interval estimation.

**Q9. What is Central Limit Theorem? State the significance of the Central Limit Theorem.**

**A9.**

* **Central Limit Theorem:**
    - States that the distribution of the sample means of a sufficiently large number of independent and identically distributed random variables will be approximately normal, regardless of the underlying distribution of the individual variables.

* **Significance:**
    - **Makes statistical inference possible:** Allows us to make inferences about the population mean even if the population distribution is not normal.
    - **Foundation for many statistical tests:** Many statistical tests, such as t-tests and z-tests, rely on the Central Limit Theorem.
    - **Improves accuracy:** As the sample size increases, the distribution of sample means becomes more and more closely approximated by a normal distribution.

**Q10. State the assumptions of the Central Limit Theorem.**

**A10.**

The Central Limit Theorem generally holds true under the following assumptions:

1. **Independence:** The random variables in the sample must be independent of each other.
2. **Identical distribution:** The random variables must be drawn from the same population distribution.
3. **Sample size:** The sample size should be sufficiently large. A common rule of thumb is that the sample size should be at least 30. However, the required sample size can vary depending on the shape of the underlying distribution.
