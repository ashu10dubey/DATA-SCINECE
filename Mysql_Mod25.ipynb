
### **Q1: What is a database? Differentiate between SQL and NoSQL databases.**  
#### **Answer:**
- A **database** is a structured collection of data that can be easily accessed, managed, and updated.
- The key differences between **SQL** and **NoSQL** databases are:

| Feature        | SQL (Relational) | NoSQL (Non-Relational) |
|---------------|----------------|----------------------|
| Data Model    | Structured (Tables) | Unstructured (JSON, Key-Value, Document) |
| Schema        | Fixed schema | Dynamic schema |
| Scalability   | Vertical scaling | Horizontal scaling |
| Transactions  | Supports ACID properties | BASE properties (Eventually Consistent) |
| Examples      | MySQL, PostgreSQL, Oracle | MongoDB, Cassandra, Redis |

### **Q2: What is DDL? Explain why CREATE, DROP, ALTER, and TRUNCATE are used with an example.**  
#### **Answer:**
- **DDL (Data Definition Language)** is used to define and manage database schema.  
- **Commands:**
  - `CREATE`: Creates tables or databases.
  - `DROP`: Deletes a table/database.
  - `ALTER`: Modifies a table structure.
  - `TRUNCATE`: Deletes all rows from a table but keeps the structure.

**Example SQL Code:**
```sql
CREATE TABLE Students (
    ID INT PRIMARY KEY,
    Name VARCHAR(100),
    Age INT
);

ALTER TABLE Students ADD COLUMN Grade VARCHAR(5);

TRUNCATE TABLE Students;

DROP TABLE Students;
```

---

### **Q3: What is DML? Explain INSERT, UPDATE, and DELETE with an example.**  
#### **Answer:**
- **DML (Data Manipulation Language)** allows modification of data within tables.
- **Commands:**
  - `INSERT`: Adds new records.
  - `UPDATE`: Modifies existing records.
  - `DELETE`: Removes records.

**Example SQL Code:**
```sql
INSERT INTO Students (ID, Name, Age, Grade) VALUES (1, 'Alice', 20, 'A');

UPDATE Students SET Grade = 'B' WHERE ID = 1;

DELETE FROM Students WHERE ID = 1;
```

### **Q4: What is DQL? Explain SELECT with an example.**  
#### **Answer:**
- **DQL (Data Query Language)** retrieves data from a database.
- The `SELECT` statement is used to fetch records.

**Example SQL Code:**
```sql
SELECT * FROM Students;

SELECT Name, Age FROM Students WHERE Grade = 'A';
```

### **Q5: Explain Primary Key and Foreign Key.**  
#### **Answer:**
- **Primary Key:** A unique identifier for a table. Cannot have NULL values.
- **Foreign Key:** A column that refers to the primary key of another table to establish a relationship.

**Example SQL Code:**
```sql
CREATE TABLE Teachers (
    TeacherID INT PRIMARY KEY,
    Name VARCHAR(100)
);

CREATE TABLE Classes (
    ClassID INT PRIMARY KEY,
    TeacherID INT,
    FOREIGN KEY (TeacherID) REFERENCES Teachers(TeacherID)
);
```

### **Q6: Write a Python code to connect MySQL to Python. Explain the cursor() and execute() method.**  
#### **Answer:**
- `cursor()`: Creates a cursor object to execute queries.
- `execute()`: Runs SQL commands.

**Python Code:**
```python
import mysql.connector

# Establish connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="password",
    database="school"
)

# Create cursor
cursor = conn.cursor()

# Execute SQL query
cursor.execute("SELECT * FROM Students")

# Fetch and print results
for row in cursor.fetchall():
    print(row)

# Close connection
cursor.close()
conn.close()
```

### **Q7: Give the order of execution of SQL clauses in an SQL query.**  
#### **Answer:**
SQL executes in the following order:
1. **FROM** – Identify the table
2. **WHERE** – Filter rows
3. **GROUP BY** – Group the result
4. **HAVING** – Filter grouped rows
5. **SELECT** – Choose columns
6. **ORDER BY** – Sort the results
7. **LIMIT** – Restrict the number of rows

**Example Query:**
```sql
SELECT Name, COUNT(*)
FROM Students
WHERE Age > 18
GROUP BY Name
HAVING COUNT(*) > 1
ORDER BY Name ASC
LIMIT 5;
```
