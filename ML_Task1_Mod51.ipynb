#### **Q1: Explain the following with an example:**
##### **A) Artificial Intelligence (AI):**  
AI refers to the simulation of human intelligence in machines that can perform tasks such as problem-solving, decision-making, and natural language processing.  
✅ **Example:** Virtual assistants like **Siri and Google Assistant** use AI to understand and respond to user queries.

##### **B) Machine Learning (ML):**  
ML is a subset of AI that enables machines to learn from data and make predictions without being explicitly programmed.  
✅ **Example:** **Spam email detection**, where an ML model learns to differentiate between spam and non-spam emails.

##### **C) Deep Learning (DL):**  
DL is a subset of ML that uses neural networks with multiple layers to extract high-level patterns from data.  
✅ **Example:** **Face recognition systems** use deep learning models to identify individuals in photos.

---

#### **Q2: What is supervised learning? List some examples of supervised learning.**  
Supervised learning is a type of machine learning where the model learns from labeled data. Each training example consists of an input and a corresponding correct output.  

✅ **Examples:**  
1. **Email spam classification** (Spam vs. Not Spam)  
2. **House price prediction** (Predicting price based on area, number of bedrooms, etc.)  
3. **Medical diagnosis** (Predicting whether a patient has a disease based on symptoms)  

---

#### **Q3: What is unsupervised learning? List some examples of unsupervised learning.**  
Unsupervised learning is a type of machine learning where the model learns from **unlabeled** data. The model identifies patterns, structures, or relationships in the data without explicit labels.  

✅ **Examples:**  
1. **Customer segmentation** (Grouping customers based on purchasing behavior)  
2. **Anomaly detection** (Detecting fraud in credit card transactions)  
3. **Topic modeling** (Identifying themes in a collection of news articles)  

---

#### **Q4: What is the difference between AI, ML, DL, and DS?**  

| **Concept** | **Definition** | **Example** |
|------------|--------------|------------|
| **Artificial Intelligence (AI)** | A broad field that enables machines to simulate human intelligence. | Self-driving cars |
| **Machine Learning (ML)** | A subset of AI that enables computers to learn from data. | Spam email classification |
| **Deep Learning (DL)** | A subset of ML that uses deep neural networks. | Image recognition |
| **Data Science (DS)** | A field that combines AI, ML, statistics, and domain knowledge to extract insights from data. | Data analysis in healthcare |

---

#### **Q5: What are the main differences between supervised, unsupervised, and semi-supervised learning?**  

| **Type** | **Definition** | **Example** |
|---------|-------------|------------|
| **Supervised Learning** | Learning from labeled data (each input has a correct output). | Predicting house prices |
| **Unsupervised Learning** | Learning from unlabeled data (finding patterns without predefined labels). | Clustering customers into segments |
| **Semi-supervised Learning** | Learning from a small amount of labeled data and a large amount of unlabeled data. | Classifying handwritten digits with only some labeled examples |

---

#### **Q6: What is train, test, and validation split? Explain the importance of each term.**  
1. **Training Set:** Used to train the model by adjusting its parameters.  
2. **Validation Set:** Used to fine-tune the model and prevent overfitting.  
3. **Test Set:** Used to evaluate the final model’s performance on unseen data.  

✅ **Importance:**  
- The training set helps the model learn.  
- The validation set ensures optimal tuning.  
- The test set measures real-world performance.  

---

#### **Q7: How can unsupervised learning be used in anomaly detection?**  
Unsupervised learning can detect anomalies by identifying data points that **deviate significantly** from normal patterns.  

✅ **Example:**  
- **Fraud detection in banking:** Unsupervised algorithms like **Isolation Forest** or **Autoencoders** can detect unusual transactions.  
- **Network security:** Identifying suspicious network activity to detect cyber threats.
