## **Q1. Explain the concept of R-squared in linear regression models. How is it calculated, and what does it represent?**  

### **Concept of R-squared (\( R^2 \))**  
R-squared (\( R^2 \)) is a statistical measure that indicates how well the independent variables explain the variance in the dependent variable in a regression model. It is also known as the **coefficient of determination**.  

### **Formula**  
$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$  
Where:  
$$( SS_{res} ) (Residual Sum of Squares) = ( \sum (y_i - \hat{y}_i)^2 )  
$$
$$( SS_{tot} ) (Total Sum of Squares) = ( \sum (y_i - \bar{y})^2 )
$$
$$( y_i ) = actual values, ( \hat{y}_i ) = predicted values, ( \bar{y} ) = mean of actual values.  $$

### **Interpretation**  
- \( R^2 = 1 \) → The model perfectly explains all variability in the data.  
- \( R^2 = 0 \) → The model explains none of the variability (i.e., predictions are no better than the mean).  
- Higher \( R^2 \) values indicate better model fit, but they do not imply causation or correctness of the model.

## **Q2. Define adjusted R-squared and explain how it differs from the regular R-squared.**  

### **Concept of Adjusted \( R^2 \)**  
Adjusted R-squared adjusts the regular \( R^2 \) for the number of predictors in the model. Unlike \( R^2 \), it penalizes the addition of irrelevant features, preventing overestimation of model performance.  

### **Formula**  
$$
R^2_{adj} = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)
$$
Where:  
- \( n \) = number of observations  
- \( p \) = number of predictors  
- \( R^2 \) = regular coefficient of determination  

### **Difference from \( R^2 \)**  
- **\( R^2 \) always increases** when adding predictors, even if they are irrelevant.  
- **\( R^2_{adj} \) only increases** if the new predictor **significantly improves the model**; otherwise, it decreases.

## **Q3. When is it more appropriate to use adjusted R-squared?**  
- When **comparing models with different numbers of predictors** (helps avoid overfitting).  
- When determining if adding more variables **genuinely improves** model performance.  
- In **multiple linear regression**, where there are many predictors.

## **Q4. What are RMSE, MSE, and MAE in the context of regression analysis? How are these metrics calculated, and what do they represent?**  

### **Mean Absolute Error (MAE)**  
$$
MAE = \frac{1}{n} \sum |y_i - \hat{y}_i|
$$  
- Measures average absolute differences between actual and predicted values.  
- Less sensitive to outliers than MSE or RMSE.  

### **Mean Squared Error (MSE)**  
$$
MSE = \frac{1}{n} \sum (y_i - \hat{y}_i)^2
$$  
- Penalizes larger errors more than MAE due to squaring.  
- Common in theoretical work but not in direct model evaluation.  

### **Root Mean Squared Error (RMSE)**  
$$
RMSE = \sqrt{MSE}
$$  
- Similar to MSE but in the original unit of the target variable.  
- More sensitive to large errors (outliers).

## **Q5. Discuss the advantages and disadvantages of using RMSE, MSE, and MAE as evaluation metrics in regression analysis.**  

| Metric  | Advantages | Disadvantages |
|---------|------------|--------------|
| **MAE** | Easy to interpret, less sensitive to outliers | Treats all errors equally, ignores squared impact |
| **MSE** | Penalizes large errors (useful for optimizing models) | Harder to interpret, sensitive to outliers |
| **RMSE** | In same units as target, emphasizes larger errors | Can be too sensitive to large errors |

## **Q6. Explain the concept of Lasso regularization. How does it differ from Ridge regularization, and when is it more appropriate to use?**  

### **Lasso Regression (\( L_1 \) Regularization)**  
Lasso (Least Absolute Shrinkage and Selection Operator) adds an absolute penalty term to the cost function, encouraging sparsity.  

$$
J(\theta) = \sum (y_i - \hat{y}_i)^2 + \lambda \sum |\theta_j|
$$  
- Forces some coefficients to **exactly zero**, performing feature selection.  
- Useful when we expect **only a few features** to be important.  

### **Difference from Ridge Regression**  
- **Ridge (\( L_2 \))**: Penalizes squared coefficients, **shrinking** them but keeping all.  
- **Lasso (\( L_1 \))**: Forces some coefficients to **zero**, removing less important features.

## **Q7. How do regularized linear models help to prevent overfitting in machine learning? Provide an example to illustrate.**  
- Regularization **adds a penalty** to large coefficients, preventing over-complex models.
- **Example**:  
  - Without regularization, a model with **many** parameters overfits.
  - With Ridge/Lasso, unnecessary parameters are **shrunk** or **removed**.

**Illustration:**  
- Model with 10 predictors, but only 3 are relevant.
- **Lasso regression** eliminates 7 irrelevant predictors.

## **Q8. Discuss the limitations of regularized linear models and explain why they may not always be the best choice for regression analysis.**  

| Limitation | Explanation |
|------------|------------|
| **Feature selection may not be optimal** | Lasso might remove important features. |
| **Does not handle non-linearity well** | Regularized linear models assume linearity. |
| **Sensitive to scaling** | Feature standardization is required. |

## **Q9. You are comparing the performance of two regression models using different evaluation metrics. Model A has an RMSE of 10, while Model B has an MAE of 8. Which model would you choose as the better performer, and why? Are there any limitations to your choice of metric?**  

- **Model A (RMSE = 10) → More sensitive to large errors.**  
- **Model B (MAE = 8) → Measures absolute differences.**  
- **Choice**:  
  - If outliers are important → **RMSE is better**.  
  - If absolute errors matter → **MAE is better**.  
  - Ideally, compare **both** to understand model behavior.

## **Q10. You are comparing the performance of two regularized linear models using different types of regularization. Model A uses Ridge regularization with a regularization parameter of 0.1, while Model B uses Lasso regularization with a regularization parameter of 0.5. Which model would you choose as the better performer, and why? Are there any trade-offs or limitations to your choice of regularization method?**  

- **Model A (Ridge, λ = 0.1) → Shrinks coefficients but keeps all.**  
- **Model B (Lasso, λ = 0.5) → Eliminates some coefficients.**  
- **Choice**:  
  - If feature selection is important → **Lasso is better**.  
  - If we need all features but reduced variance → **Ridge is better**.  

**Trade-offs**:  
- Lasso may remove useful features.  
- Ridge does not perform feature selection.
