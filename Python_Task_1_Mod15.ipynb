Q1. Create a function which will take a list as an argument and return the product of all the numbers
after creating a flat list.
Use the below-given list as an argument for your function.
list1 = [1,2,3,4, [44,55,66, True], False, (34,56,78,89,34), {1,2,3,3,2,1}, {1:34, "key2": [55, 67, 78, 89], 4: (45,
22, 61, 34)}, [56, 'data science'], 'Machine Learning']
Note: you must extract numeric keys and values of the dictionary also.

```python
from collections.abc import Iterable

def flatten_and_multiply(lst):
    flat_list = []

    def flatten(element):
        """Recursively extract numeric values from various iterable structures."""
        if isinstance(element, dict):
            for key, value in element.items():
                if isinstance(key, (int, float)):  # Extract numeric keys
                    flat_list.append(key)
                flatten(value)  # Extract values
        elif isinstance(element, (list, set, tuple)):
            for item in element:
                flatten(item)
        elif isinstance(element, (int, float)):  # Extract only numbers
            flat_list.append(element)

    flatten(lst)  # Flatten the nested structures

    # Compute product of all numbers in flat_list
    product = 1
    for num in flat_list:
        product *= num

    return product

# Given list
list1 = [
    1, 2, 3, 4,
    [44, 55, 66, True], False,
    (34, 56, 78, 89, 34),
    {1, 2, 3, 3, 2, 1},
    {1: 34, "key2": [55, 67, 78, 89], 4: (45, 22, 61, 34)},
    [56, 'data science'], 'Machine Learning'
]

# Call the function
result = flatten_and_multiply(list1)
print("Product of all numbers:", result)
```

Q2. Write a python program for encrypting a message sent to you by your friend. The logic of encryption
should be such that, for a the output should be z. For b, the output should be y. For c, the output should
be x respectively. Also, the whitespace should be replaced with a dollar sign. Keep the punctuation
marks unchanged.
Input Sentence: I want to become a Data Scientist.
Encrypt the above input sentence using the program you just created.
Note: Convert the given input sentence into lowercase before encrypting. The final output should be
lowercase.

```python
import string

def encrypt_message(message):
    alphabet = string.ascii_lowercase  # 'abcdefghijklmnopqrstuvwxyz'
    reverse_alphabet = alphabet[::-1]  # 'zyxwvutsrqponmlkjihgfedcba'

    encryption_dict = {alphabet[i]: reverse_alphabet[i] for i in range(26)}
    encryption_dict[' '] = '$'  # Replace spaces with $

    encrypted_message = ''.join(encryption_dict.get(char, char) for char in message.lower())

    return encrypted_message

# Given sentence
input_sentence = "I want to become a Data Scientist."

# Encrypt the message
encrypted_output = encrypt_message(input_sentence)
print("Encrypted Message:", encrypted_output)
```
