### **Q1. What is an activation function in the context of artificial neural networks?**  
An **activation function** introduces non-linearity into a neural network, allowing it to learn **complex patterns** beyond simple linear relationships. It decides whether a neuron should be activated based on the input it receives. Without activation functions, neural networks would be equivalent to linear regression models.  

---

### **Q2. What are some common types of activation functions used in neural networks?**  
Some widely used activation functions include:  
1. **Sigmoid (\(\sigma(x)\))** – Outputs values between 0 and 1.  
2. **Hyperbolic Tangent (tanh)** – Outputs values between -1 and 1.  
3. **ReLU (Rectified Linear Unit)** – Outputs \( \max(0, x) \).  
4. **Leaky ReLU** – A variant of ReLU that allows small negative values.  
5. **Softmax** – Converts outputs into probabilities, used for classification.  
6. **Swish & GELU** – Advanced activation functions used in deep learning models.  

---

### **Q3. How do activation functions affect the training process and performance of a neural network?**  
Activation functions impact:  
- **Non-linearity** – Allowing the network to learn complex patterns.  
- **Gradient Flow** – Some activations can cause vanishing/exploding gradients.  
- **Convergence Speed** – Faster convergence with functions like ReLU.  
- **Output Range** – Determines stability and interpretability of outputs.  

For example, **ReLU speeds up training**, while **sigmoid/tanh may slow it down** due to saturation.  

---

### **Q4. How does the sigmoid activation function work? What are its advantages and disadvantages?**  
The **sigmoid function** is defined as:  
$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$  
- **Advantages:**  
  - Outputs between **0 and 1**, making it interpretable as a probability.  
  - Useful in binary classification tasks.  
- **Disadvantages:**  
  - **Vanishing Gradient Problem** – When inputs are too large or too small, gradients become close to **zero**, slowing learning.  
  - **Not Zero-Centered** – Causes inefficient weight updates.  
  - **Computationally Expensive** – Involves exponentiation.  

---

### **Q5. What is the rectified linear unit (ReLU) activation function? How does it differ from the sigmoid function?**  
The **ReLU function** is defined as:  
\[
f(x) = \max(0, x)
\]  
- **Key Differences from Sigmoid:**  
  - **Faster convergence** – No exponentiation.  
  - **No vanishing gradients for positive inputs**.  
  - **Zero-centered for negative inputs** but can suffer from **"dying ReLU"**, where neurons output zero for all inputs.  

---

### **Q6. What are the benefits of using the ReLU activation function over the sigmoid function?**  
1. **Avoids Vanishing Gradient** – Unlike sigmoid, ReLU doesn’t squash large values.  
2. **Faster Computation** – Simple max(0, x) operation.  
3. **Sparse Activation** – Some neurons output zero, reducing unnecessary computations.  
4. **Better Performance in Deep Networks** – Helps train deep models efficiently.  

---

### **Q7. Explain the concept of "leaky ReLU" and how it addresses the vanishing gradient problem.**  
**Leaky ReLU** modifies ReLU to allow small negative values instead of zero:  
$$
f(x) =
\begin{cases}
    x, & \text{if } x > 0 \\
    \alpha x, & \text{if } x \leq 0 \text{ (where } 0 < \alpha < 1)
\end{cases}
$$  
- **Solves "dying ReLU"** problem, where neurons get stuck with zero output.  
- **Retains benefits of ReLU** while ensuring small gradients exist for negative inputs.  

---

### **Q8. What is the purpose of the softmax activation function? When is it commonly used?**  
Softmax converts raw output scores into probabilities:  
$$
\sigma(z_i) = \frac{e^{z_i}}{\sum_{j} e^{z_j}}
$$  
- **Purpose:** Converts logits into a probability distribution.  
- **Used in:**  
  - **Multi-class classification** (e.g., Image Classification).  
  - **Last layer of a neural network** for categorical predictions.  

---

### **Q9. What is the hyperbolic tangent (tanh) activation function? How does it compare to the sigmoid function?**  
The **tanh function** is defined as:  
$$
tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
$$  
- **Key Differences from Sigmoid:**  
  - Outputs values between **-1 and 1** instead of **0 and 1**.  
  - **Zero-centered**, making weight updates more efficient.  
  - Still suffers from **vanishing gradient** for large values.  

**Use case:** Works better than sigmoid when **negative values matter** (e.g., RNNs).
