### **Q1: What are missing values in a dataset? Why is it essential to handle missing values? Name some algorithms that are not affected by missing values.**  

**Answer:**  
Missing values in a dataset refer to the absence of data in one or more fields. These missing values can occur due to various reasons such as data corruption, human error, or incomplete data collection.  

**Importance of Handling Missing Values:**  
- Prevents biased analysis and incorrect model predictions.  
- Enhances data integrity and reliability.  
- Some machine learning algorithms cannot handle missing values and may fail.  

**Algorithms Not Affected by Missing Values:**  
- **Decision Trees (e.g., Random Forest, XGBoost)**
- **K-Nearest Neighbors (KNN) (if using distance-weighted imputation)**
- **Naïve Bayes (as it works with probabilities, missing values are ignored)**

### **Q2: List down techniques used to handle missing data. Give an example of each with Python code.**  

**Answer:**  
There are several techniques to handle missing data:  

1. **Deletion Methods:**  
   - Remove rows or columns with missing values (only when missing data is insignificant).  
   ```python
   import pandas as pd
   
   df = pd.DataFrame({"A": [1, 2, None], "B": [4, None, 6]})
   df_cleaned = df.dropna()  # Drop rows with missing values
   print(df_cleaned)
   ```  

2. **Imputation Methods:**  
   - **Mean/Median/Mode Imputation** (For numerical data):  
   ```python
   df.fillna(df.mean(), inplace=True)  # Replacing missing values with column mean
   ```  
   - **Forward or Backward Fill (For time-series data)**  
   ```python
   df.fillna(method='ffill', inplace=True)  # Forward fill
   df.fillna(method='bfill', inplace=True)  # Backward fill
   ```  

3. **Predictive Imputation (Using ML algorithms like KNN or Regression):**  
   ```python
   from sklearn.impute import KNNImputer
   imputer = KNNImputer(n_neighbors=2)
   df_filled = imputer.fit_transform(df)
   ```

### **Q3: Explain imbalanced data. What will happen if imbalanced data is not handled?**  

**Answer:**  
Imbalanced data occurs when the distribution of classes in a dataset is highly skewed (e.g., 90% Class A and 10% Class B).  

**Consequences of Not Handling Imbalanced Data:**  
- **Poor Model Performance**: The model will be biased toward the majority class.  
- **Misleading Accuracy**: High accuracy does not mean good predictive power if the minority class is ignored.  
- **Increased False Negatives**: Rare events (e.g., fraud detection, disease diagnosis) may not be detected.  

**Solutions:**  
- Use resampling techniques (Oversampling, Undersampling).  
- Use cost-sensitive algorithms or class-weighted models.  
- Use synthetic data augmentation techniques like **SMOTE**.
