### **Q3. Suppose you are conducting a study to examine the relationship between the number of hours of exercise per week and body mass index (BMI) in a sample of adults. You collected data on both variables for 50 participants. Calculate the Pearson correlation coefficient and the Spearman's rank correlation between these two variables and compare the results.**

You're asked to calculate *both* Pearson and Spearman correlations.  This is important because they measure different types of relationships:

*   **Pearson Correlation:** Measures the *linear* relationship between two *continuous* variables. It assumes that the data is approximately normally distributed.  It's sensitive to outliers.
*   **Spearman Rank Correlation:** Measures the *monotonic* relationship between two variables.  A monotonic relationship means that as one variable increases, the other *tends* to either increase or decrease, but not necessarily at a constant rate.  Spearman is non-parametric (doesn't assume normality) and is less sensitive to outliers.

**Calculations (Manual - Tedious but Illustrative):**

1.  **Organize your data:** Create two lists or arrays: one for exercise hours and one for BMI.

2.  **Pearson Correlation:**
    *   Calculate the mean and standard deviation of both variables.
    *   For each participant, calculate the z-score for exercise hours and the z-score for BMI.  The z-score is `(value - mean) / standard deviation`.
    *   Multiply the z-scores for each participant.
    *   Sum the products of the z-scores.
    *   Divide the sum by `n-1` (where `n` is the number of participants, 50 in this case).

3.  **Spearman Rank Correlation:**
    *   Rank the exercise hours from lowest to highest.  Assign rank 1 to the lowest value, 2 to the next lowest, and so on.  Handle ties by assigning the average rank.  Do the same for BMI.
    *   Calculate the difference between the ranks for each participant.
    *   Square the differences.
    *   Sum the squared differences.
    *   Use the Spearman rank correlation formula:

    ```
    rs = 1 - (6 * Σd^2) / (n * (n^2 - 1))
    ```

    Where `d` is the difference in ranks and `n` is the number of participants.

**Python Code (Recommended):**

```python
from scipy.stats import pearsonr, spearmanr

exercise_hours = [ /* your exercise hours data here */ ]  # Replace with your actual data
bmi = [ /* your BMI data here */ ]  # Replace with your actual data

pearson_corr, _ = pearsonr(exercise_hours, bmi)
spearman_corr, _ = spearmanr(exercise_hours, bmi)

print(f"Pearson Correlation: {pearson_corr}")
print(f"Spearman Correlation: {spearman_corr}")
```

**Comparing the Results:**

*   If both correlations are similar, it suggests a linear relationship.
*   If Spearman is higher than Pearson, it suggests a monotonic relationship, but the relationship might not be strictly linear.
*   If Pearson is close to zero but Spearman is not, it suggests that there might be a relationship, but not a linear one.
*   The *p-value* (the second value returned by `pearsonr` and `spearmanr`) indicates the statistical significance of the correlation. A low p-value (typically less than 0.05) suggests that the correlation is statistically significant (unlikely due to chance).

###Q5:image.png
This problem requires a **contingency table** and a **Chi-Square test** to determine if there's a statistically significant relationship between age groups and soft drink preference.  Here's how to approach it:

**1. Group the Ages:**

Since you have individual ages, you need to create age groups (categories).  There's no single right way, but you should aim for meaningful groupings based on the context.  For this example, let's use:

*   **18-25 (Young Adult)**
*   **26-35 (Adult)**
*   **36-45 (Middle-Aged Adult)**

**2. Create the Contingency Table:**

Count how many people in each age group prefer each soft drink.

| Age Group        | Coke | Pepsi | Mountain Dew | Total |
|-----------------|------|-------|--------------|-------|
| 18-25 (Young Adult) | 2    | 0     | 0            | 2     |
| 26-35 (Adult)   | 1    | 1     | 1            | 3     |
| 36-45 (Middle-Aged Adult) | 0    | 1     | 1            | 2     |
| **Total**       | 3    | 2     | 2            | 7     |

**3. Perform the Chi-Square Test:**

The Chi-Square test determines if the observed distribution of preferences is significantly different from what you'd expect if there were *no* relationship between age and preference.

*   **State the Hypotheses:**
    *   **Null Hypothesis (H0):** Age and soft drink preference are independent (no relationship).
    *   **Alternative Hypothesis (H1):** Age and soft drink preference are not independent (there is a relationship).

*   **Calculate Expected Frequencies:** For each cell in the table, calculate the expected frequency assuming independence.
    *   `Expected Frequency = (Row Total * Column Total) / Grand Total`

    For example, the expected frequency for Young Adults preferring Coke is:  `(2 * 3) / 7 = 0.86`

| Age Group        | Coke (Expected) | Pepsi (Expected) | Mountain Dew (Expected) | Total |
|-----------------|---------------|----------------|-----------------------|-------|
| 18-25 (Young Adult) | 0.86          | 0.57           | 0.57                  | 2     |
| 26-35 (Adult)   | 1.29          | 0.86           | 0.86                  | 3     |
| 36-45 (Middle-Aged Adult) | 0.86          | 0.57           | 0.57                  | 2     |
| **Total**       | 3             | 2              | 2                     | 7     |

*   **Calculate the Chi-Square Statistic:**

    ```
    χ² = Σ [(Observed Frequency - Expected Frequency)² / Expected Frequency]
    ```

    Sum this calculation for each cell in the table.

    For example, for Young Adults preferring Coke: `(2 - 0.86)² / 0.86 = 1.54`

    Do this for all cells and sum the results.

*   **Determine Degrees of Freedom (df):**

    ```
    df = (number of rows - 1) * (number of columns - 1)
    ```

    In this case, df = (3-1) * (3-1) = 4

*   **Find the Critical Value or p-value:**

    *   **Critical Value:** Using a Chi-Square distribution table and your chosen significance level (alpha, usually 0.05), find the critical value corresponding to your degrees of freedom. If your calculated χ² is greater than the critical value, you reject the null hypothesis.
    *   **p-value:** Use a statistical calculator or software (like R or Python with `scipy.stats`) to find the p-value associated with your calculated χ².  If the p-value is less than your significance level (alpha), you reject the null hypothesis.

**4. Conclusion:**

*   If you reject the null hypothesis, there is evidence of a statistically significant relationship between age and soft drink preference.
*   If you fail to reject the null hypothesis, there is not enough evidence to suggest a relationship between age and soft drink preference.

**Python Code (Using `scipy.stats`):**

```python
import numpy as np
from scipy.stats import chi2_contingency

observed = np.array([[2, 0, 0], [1, 1, 1], [0, 1, 1]])  # Your contingency table data

chi2, p, dof, expected = chi2_contingency(observed)

print(f"Chi-square statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of freedom: {dof}")
print("Expected frequencies:\n", expected)
```

This code will give you the Chi-square statistic, the p-value, the degrees of freedom, and the expected frequencies.  You can then interpret the p-value as described above.

**Important Note:**  With a small sample size (like 7 in this example), the Chi-Square test might not be very reliable.  A larger sample size would provide more robust results. Additionally, the way age was grouped can influence the results.  Consider the context and experiment with different age groupings if necessary.
