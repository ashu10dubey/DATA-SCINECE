### **Q1. What is the difference between Ordinal Encoding and Label Encoding? Provide an example of when you might choose one over the other.**  
**Answer:**  
### **Difference Between Ordinal Encoding and Label Encoding**

| Feature             | **Ordinal Encoding**                     | **Label Encoding**                      |
|---------------------|-------------------------------------|-------------------------------------|
| **Definition**         | Assigns integer values based on order/rank. | Assigns unique integers arbitrarily. |
| **When to Use**       | When categorical values have a meaningful order (e.g., Education level: High School < Bachelor’s < Master’s). | When there is no inherent order (e.g., Car brands: Tesla, BMW, Ford). |
| **Example**           | **Education Level:** High School → 1, Bachelor's → 2, Master's → 3, PhD → 4 | **Car Brands:** Tesla → 0, BMW → 1, Ford → 2 |

### **Example of When to Choose One Over the Other**

#### **1️⃣ Ordinal Encoding Example (When Order Matters)**
Suppose you have a feature **"Customer Satisfaction"** with the categories:  
- **Very Dissatisfied → 1**  
- **Dissatisfied → 2**  
- **Neutral → 3**  
- **Satisfied → 4**  
- **Very Satisfied → 5**  

Since there is a meaningful order in customer satisfaction levels, **Ordinal Encoding** is the correct choice.

#### **2️⃣ Label Encoding Example (When Order Doesn't Matter)**
Consider a feature **"Car Brand"** with the categories:  
- **Tesla → 0**  
- **BMW → 1**  
- **Ford → 2**  

Since there is **no inherent ranking** among car brands, Label Encoding is appropriate.

### **Key Takeaways**
✅ Use **Ordinal Encoding** for ordered categorical data.  
✅ Use **Label Encoding** for nominal categorical data (no natural order).  
✅ Be cautious: Using **Label Encoding** on ordinal data can mislead ML models.

### **Q2. Explain how Target Guided Ordinal Encoding works and provide an example of when you might use it in a machine learning project.**  
**Answer:**  
**Target Guided Ordinal Encoding** assigns numerical values to categories based on their relationship with the target variable (dependent variable).  

**Steps:**  
1. Compute the mean (or median) of the target variable for each category.  
2. Assign ordinal values based on the sorted mean target values.  

**Example:**  
If you are predicting **customer churn (0 = No, 1 = Yes)** and the **contract type** feature has three categories:  

| Contract Type  | Avg Churn Rate |
|---------------|---------------|
| Monthly       | 0.60          |
| Yearly        | 0.30          |
| Two-Year      | 0.10          |

Encoding:  
- Monthly → 3  
- Yearly → 2  
- Two-Year → 1  

**When to Use:**  
- When there is a correlation between categorical features and the target variable.  
- Useful in predictive modeling to improve performance.

### **Covariance: Definition & Importance in Statistical Analysis**  

#### **Definition:**  
Covariance is a statistical measure that quantifies the degree to which two variables change together. If two variables increase or decrease simultaneously, they have a **positive covariance**. If one increases while the other decreases, they have a **negative covariance**. A covariance close to zero indicates little to no relationship between the variables.  

#### **Importance in Statistical Analysis:**  
1. **Understanding Relationships:** Covariance helps determine whether two variables are related and how they move together.  
2. **Feature Selection in Machine Learning:** High covariance between features may indicate redundancy, leading to dimensionality reduction techniques like **Principal Component Analysis (PCA)**.  
3. **Portfolio Management:** In finance, covariance is used to assess the relationship between different assets, aiding in risk management.  
4. **Correlation Calculation:** Covariance is the basis for calculating **correlation**, which standardizes the relationship between variables.  

#### **Formula for Covariance:**  
For two variables **X** and **Y**, the covariance is calculated as:  

\[
Cov(X, Y) = \frac{\sum (X_i - \bar{X}) (Y_i - \bar{Y})}{n}
\]

Where:  
- \( X_i \), \( Y_i \) are individual data points.  
- \( \bar{X} \), \( \bar{Y} \) are the means of X and Y.  
- \( n \) is the total number of data points.

### **Q4. For a dataset with the following categorical variables: Color (red, green, blue), Size (small, medium, large), and Material (wood, metal, plastic), perform label encoding using Python's scikit-learn library. Show your code and explain the output.**  
**Answer (Python Code):**
Here’s how you can perform **Label Encoding** on the categorical variables **Color, Size, and Material** using Python's `sklearn.preprocessing.LabelEncoder`:

---

### **Python Code:**
```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# Creating the dataset
data = {
    'Color': ['red', 'green', 'blue', 'green', 'red', 'blue'],
    'Size': ['small', 'medium', 'large', 'medium', 'small', 'large'],
    'Material': ['wood', 'metal', 'plastic', 'metal', 'wood', 'plastic']
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Initialize LabelEncoder
encoder = LabelEncoder()

# Apply Label Encoding to each categorical column
df_encoded = df.copy()
for column in df.columns:
    df_encoded[column] = encoder.fit_transform(df[column])

# Display the original and encoded DataFrame
print("Original DataFrame:\n", df)
print("\nLabel Encoded DataFrame:\n", df_encoded)
```

---

### **Explanation:**
1. We created a dataset with three categorical variables: **Color, Size, and Material**.
2. We converted it into a **pandas DataFrame**.
3. We initialized `LabelEncoder` and applied it **column-wise** to encode the categorical values into numerical values.
4. The transformed **encoded DataFrame** replaces categorical values with unique numeric identifiers.

---

### **Expected Output:**
#### **Original DataFrame:**
| Color | Size  | Material |
|-------|-------|---------|
| red   | small | wood    |
| green | medium | metal  |
| blue  | large  | plastic |
| green | medium | metal  |
| red   | small | wood    |
| blue  | large  | plastic |

#### **Label Encoded DataFrame:**
| Color | Size | Material |
|-------|------|---------|
| 2     | 2    | 2       |
| 1     | 1    | 0       |
| 0     | 0    | 1       |
| 1     | 1    | 0       |
| 2     | 2    | 2       |
| 0     | 0    | 1       |

---

### **How Label Encoding Works:**
- The encoder assigns **numeric values** based on alphabetical order of the categories.
- For **Color**: `blue → 0`, `green → 1`, `red → 2`
- For **Size**: `large → 0`, `medium → 1`, `small → 2`
- For **Material**: `metal → 0`, `plastic → 1`, `wood → 2`

---

### **Key Takeaways:**
- Label Encoding is useful for **ordinal** and **nominal** categorical variables.
- It works best when the **categorical values** do not imply any meaningful order.
- For some ML models, label encoding may introduce unintended relationships. In such cases, **One-Hot Encoding** is a better choice.

### **Q5. Calculate the covariance matrix for the following variables in a dataset: Age, Income, and Education Level. Interpret the results.**  
**Answer (Python Code):**
```python
import numpy as np

# Sample data (Age, Income, Education Level)
data = np.array([[25, 50000, 2],
                 [30, 60000, 3],
                 [35, 70000, 4],
                 [40, 80000, 4],
                 [45, 90000, 5]])

# Compute covariance matrix
cov_matrix = np.cov(data, rowvar=False)
print(cov_matrix)
```

**Interpretation:**  
- A high positive value between Age & Income suggests they increase together.  
- A low or near-zero value between Income & Education means little correlation.  
- A negative value (if any) suggests an inverse relationship.

### **Q7. You are analyzing a dataset with two continuous variables, "Temperature" and "Humidity", and two categorical variables, "Weather Condition" (Sunny/Cloudy/Rainy) and "Wind Direction" (North/South/ East/West). Calculate the covariance between each pair of variables and interpret the results.**  
**Answer (Python Code):**
```python
import pandas as pd

# Sample Data
data = {'Temperature': [30, 25, 28, 35, 32],
        'Humidity': [70, 80, 65, 60, 75],
        'Weather Condition': ['Sunny', 'Cloudy', 'Rainy', 'Sunny', 'Cloudy'],
        'Wind Direction': ['North', 'South', 'East', 'West', 'North']}

df = pd.DataFrame(data)

# Encoding Categorical Variables
df['Weather Condition'] = df['Weather Condition'].astype('category').cat.codes
df['Wind Direction'] = df['Wind Direction'].astype('category').cat.codes

# Compute Covariance Matrix
cov_matrix = df.cov()
print(cov_matrix)
```

**Interpretation:**  
- A positive covariance between Temperature & Humidity means both increase together.  
- A near-zero covariance between Weather Condition & Wind Direction suggests no strong relationship.  
- If Wind Direction and Humidity show a negative covariance, it indicates opposing trends.
