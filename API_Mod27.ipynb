## **Q1: What is an API? Give an example where an API is used in real life.**  
### **Answer:**  
- **API (Application Programming Interface)** is a set of rules that allow different software applications to communicate with each other.  
- **Example:**  
  - **Google Maps API**: Many apps like Uber, Zomato, and Swiggy use Google Maps API to fetch location data, navigation routes, and real-time traffic information.  

---

## **Q2: Advantages and Disadvantages of using API**  
### **Answer:**  
### **✅ Advantages:**  
1. **Automation:** Reduces manual intervention by automating tasks.  
2. **Reusability:** Enables reusing functionalities in different applications.  
3. **Scalability:** Helps in building scalable applications efficiently.  
4. **Security:** Data exchange occurs in a controlled manner with authentication mechanisms.  
5. **Integration:** Connects different applications, databases, or systems.  

### **❌ Disadvantages:**  
1. **Security Risks:** Improper API implementation can lead to vulnerabilities.  
2. **Maintenance Overhead:** APIs require regular updates and monitoring.  
3. **Dependency Issues:** If an API provider changes or discontinues an API, it affects dependent applications.  
4. **Rate Limits:** Many APIs impose restrictions on request frequency, affecting performance.

## **Q3: What is a Web API? Differentiate between API and Web API.**  
### **Answer:**  
- A **Web API** is an API that is accessible over the **internet** using HTTP/HTTPS protocols. It allows applications to communicate over the web.  

### **Difference Between API and Web API:**  

| Feature  | API  | Web API  |
|----------|------|---------|
| Definition | A set of functions that allow applications to interact | A type of API accessible via HTTP |
| Communication | Can work within local systems | Works over the web (HTTP/HTTPS) |
| Usage | Used in databases, operating systems, libraries | Used in web applications and cloud services |
| Example | Windows API, Java API | RESTful API, Google Maps API |

## **Q4: Explain REST and SOAP Architecture. Mention shortcomings of SOAP.**  
### **Answer:**  
### **🌐 REST (Representational State Transfer) Architecture:**  
- A lightweight architecture that uses standard HTTP methods (GET, POST, PUT, DELETE) to interact with resources.  
- Uses JSON/XML as data format.  
- Stateless, meaning each request is independent.  

### **🌍 SOAP (Simple Object Access Protocol) Architecture:**  
- A protocol that allows structured messaging via XML.  
- Heavily dependent on **WS-Security** for authentication.  
- Uses **WSDL (Web Services Description Language)** to describe operations.  

### **🚫 Shortcomings of SOAP:**  
1. **Overhead:** SOAP messages are bulky due to XML.  
2. **Complexity:** Requires additional specifications (e.g., WS-Security, WSDL).  
3. **Slower Performance:** More processing required compared to REST.
