### **Q1. What are the three measures of central tendency?**  
The three measures of central tendency are:  
1. **Mean** – The average of a dataset.  
2. **Median** – The middle value of an ordered dataset.  
3. **Mode** – The most frequently occurring value in the dataset.  

---

### **Q2. Difference between Mean, Median, and Mode and their usage**  
- **Mean**: The arithmetic average, used when data is symmetric and has no extreme values.  
- **Median**: The middle value, used when data has outliers or is skewed.  
- **Mode**: The most frequent value, used in categorical data analysis.  

Example: For the dataset **[2, 3, 5, 8, 8, 10, 12]**  
- Mean = (2+3+5+8+8+10+12)/7 = **6.86**  
- Median = **8** (middle value)  
- Mode = **8** (most frequent value)  

---

### **Q3. Measures of central tendency for given height data**  
Dataset: **[178,177,176,177,178.2,178,175,179,180,175,178.9,176.2,177,172.5,178,176.5]**  
- **Mean** = (Sum of values) / (Number of values) = **177.04**  
- **Median** = **177** (middle value after sorting)  
- **Mode** = **178** (most frequent value)

### **Q4. Standard Deviation of given data**  
Formula:  
\[
\sigma = \sqrt{\frac{\sum (x_i - \bar{x})^2}{N}}
\]
where \( x_i \) are data points, \( \bar{x} \) is the mean, and \( N \) is the number of data points.  

Calculated standard deviation: **1.99**  

---

### **Q5. Measures of Dispersion (Range, Variance, Standard Deviation)**  
- **Range**: Difference between the maximum and minimum values.  
- **Variance**: Measures the average squared deviation from the mean.  
- **Standard Deviation**: Square root of variance, indicating data spread.  

Example: Dataset **[2, 4, 6, 8, 10]**  
- Range = **10 - 2 = 8**  
- Variance = **8**  
- Standard Deviation = **2.83**

### **Q8. What is skewness in data?**  
Skewness measures the asymmetry of a dataset’s distribution.  
- **Right skew (positive skew)**: Tail on the right side, mean > median.  
- **Left skew (negative skew)**: Tail on the left side, mean < median.  

---

### **Q9. Median position in Right-Skewed Data**  
For right-skewed data:  
**Mean > Median > Mode**  

---

### **Q10. Covariance vs. Correlation**  
- **Covariance**: Measures the direction of the relationship between two variables.  
- **Correlation**: Measures both strength and direction of the relationship (ranges from -1 to +1).  

---

### **Q11. Formula for Sample Mean**  
\[
\bar{x} = \frac{\sum x_i}{N}
\]  
Example: Dataset **[10, 15, 20]**  
\[
\bar{x} = \frac{10+15+20}{3} = 15
\]  

---

### **Q12. Relationship of Central Tendency in Normal Distribution**  
For a normal distribution:  
**Mean = Median = Mode**  

---

### **Q13. Covariance vs. Correlation (again for clarity)**  
- **Covariance**: Unbounded, depends on scale.  
- **Correlation**: Standardized, always between -1 and 1.  

---

### **Q14. Effect of Outliers on Central Tendency and Dispersion**  
- **Mean**: Affected by outliers (can shift significantly).  
- **Median**: Less affected.  
- **Mode**: Not affected unless the outlier is frequent.  
- **Standard Deviation**: Increases due to extreme values.  

Example:  
Dataset **[10, 12, 14, 100]**  
- Mean = **34** (inflated by 100)  
- Median = **13** (resistant to outlier)  
- Mode = **No mode**  
- Standard deviation = **High due to 100**
