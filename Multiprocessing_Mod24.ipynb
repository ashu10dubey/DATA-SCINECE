### **Q1: What is multiprocessing in Python? Why is it useful?**  
**Answer:**  
Multiprocessing in Python is a technique that allows a program to run multiple processes concurrently, utilizing multiple CPU cores. It is useful because it enables **parallel execution**, making it ideal for CPU-bound tasks (e.g., mathematical computations, image processing, etc.). Unlike **multithreading**, which is limited by Python's Global Interpreter Lock (GIL), **multiprocessing can execute tasks truly in parallel**, leading to significant performance improvements.  

---

### **Q2: What are the differences between multiprocessing and multithreading?**  
**Answer:**  

| Feature            | Multiprocessing | Multithreading |
|--------------------|----------------|---------------|
| **Execution**     | Runs multiple processes independently | Runs multiple threads within a single process |
| **Concurrency**   | Achieves **true parallelism** | Uses concurrency but affected by GIL |
| **Best for**      | CPU-bound tasks (e.g., numerical computations) | I/O-bound tasks (e.g., file operations, network requests) |
| **Memory Usage**  | Each process has its own memory space | Threads share the same memory space |
| **Overhead**      | Higher due to process creation | Lower due to shared memory |
| **Communication** | Requires inter-process communication (IPC) | Easier as threads share memory |

### **Q3: Write a Python code to create a process using the multiprocessing module.**  

```python
import multiprocessing

def print_message():
    print("Hello from the child process!")

if __name__ == "__main__":
    process = multiprocessing.Process(target=print_message)
    process.start()  # Start the process
    process.join()   # Wait for process to complete
    print("Main process finished execution.")
```

**Output:**  
```
Hello from the child process!
Main process finished execution.
```

### **Q4: What is a multiprocessing pool in Python? Why is it used?**  
**Answer:**  
A **multiprocessing pool** in Python is a way to create a pool of worker processes that can execute tasks in parallel. It is used to efficiently distribute work among multiple processes while managing them easily.  

**Why use a multiprocessing pool?**  
- **Efficient parallel execution**: Automatically distributes tasks among available CPU cores.  
- **Resource management**: Limits the number of active processes, preventing system overload.  
- **Easy parallelism**: Provides built-in methods (`map`, `apply`, `starmap`) for task execution.

### **Q5: How can we create a pool of worker processes in Python using the multiprocessing module?**  

```python
import multiprocessing

def square(n):
    return n * n

if __name__ == "__main__":
    numbers = [1, 2, 3, 4, 5]
    pool = multiprocessing.Pool(processes=3)  # Create a pool with 3 worker processes
    results = pool.map(square, numbers)  # Apply function to each element
    pool.close()  # Close the pool
    pool.join()   # Wait for all processes to finish
    print(results)
```

**Output:**  
```
[1, 4, 9, 16, 25]
```

### **Q6: Write a Python program to create 4 processes, each process should print a different number using the multiprocessing module in Python.**  

```python
import multiprocessing

def print_number(n):
    print(f"Process {multiprocessing.current_process().name} prints {n}")

if __name__ == "__main__":
    processes = []
    for i in range(4):
        process = multiprocessing.Process(target=print_number, args=(i,))
        processes.append(process)
        process.start()  # Start process

    for process in processes:
        process.join()  # Wait for all processes to complete
```

**Output (order may vary due to parallel execution):**  
```
Process Process-1 prints 0
Process Process-2 prints 1
Process Process-3 prints 2
Process Process-4 prints 3
```

This demonstrates how multiple processes can execute simultaneously and independently using Python’s **multiprocessing module**. 🚀
